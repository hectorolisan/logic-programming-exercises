# Visualización · Dibujos y funciones

Figuras geométricas, gráficos de calificaciones y gráficas de funciones. El libro original usaba el entorno PythonG; aquí se dibuja con `matplotlib`.

## 1. Cuadrado y circunferencias concéntricas

Dibujar un cuadrado con esquinas (100, 100) y (900, 900), y tres circunferencias concéntricas de radios 100, 200 y 300.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle

fig, (a, b) = plt.subplots(1, 2, figsize=(8, 4))
a.add_patch(Rectangle((100, 100), 800, 800, fill=False))
a.set(xlim=(0, 1000), ylim=(0, 1000), aspect="equal")
for r in (100, 200, 300):
    b.add_patch(Circle((0, 0), r, fill=False))
b.set(xlim=(-350, 350), ylim=(-350, 350), aspect="equal")
plt.show()
```

## 2. Diana con ejes y letras

Las tres circunferencias con dos diámetros perpendiculares y las letras a, b, c y d en sus extremos; después, con líneas amarilla y magenta, circunferencias verdes y azules y letras negras.

```python
fig, ejes = plt.subplots(1, 2, figsize=(8, 4))
for eje, color_lineas, color_circulos in zip(ejes, (("black", "black"), ("yellow", "magenta")), (("black",) * 3, ("green", "blue", "green"))):
    eje.plot([-300, 300], [0, 0], color=color_lineas[0])
    eje.plot([0, 0], [-300, 300], color=color_lineas[1])
    for r, c in zip((100, 200, 300), color_circulos):
        eje.add_patch(Circle((0, 0), r, fill=False, color=c))
    for letra, (x, y) in zip("abcd", ((0, 320), (320, 0), (0, -340), (-340, 0))):
        eje.text(x, y, letra, color="black", ha="center", va="center")
    eje.set(xlim=(-370, 370), ylim=(-370, 370), aspect="equal")
    eje.axis("off")
plt.show()
```

## 3. Gráficos de calificaciones

Pedir el porcentaje de suspensos, aprobados, notables y sobresalientes y dibujar un gráfico de tarta; después, pedir el número de alumnos de cada calificación (convirtiéndolo en porcentajes) y dibujar la tarta y un gráfico de barras.

```python
calificaciones = ["Sus", "Apr", "Not", "Sob"]
porcentajes = [float(input(f"% de {c}: ")) for c in calificaciones]
cantidades = [int(input(f"Número de {c}: ")) for c in calificaciones]
total = sum(cantidades)
fig, (a, b, c) = plt.subplots(1, 3, figsize=(12, 4))
a.pie(porcentajes, labels=calificaciones, autopct="%1.0f%%")
b.pie([100 * n / total for n in cantidades], labels=calificaciones, autopct="%1.0f%%")
c.bar(calificaciones, [100 * n / total for n in cantidades])
c.set_ylabel("%")
plt.show()
```

## 4. Seno y coseno en un intervalo

Pedir un intervalo y dibujar la función coseno en él; después, seno y coseno a la vez en colores distintos.

```python
import numpy as np

a_ini, b_fin = float(input("Desde: ")), float(input("Hasta: "))
x = np.linspace(a_ini, b_fin, 200)
fig, (izq, der) = plt.subplots(1, 2, figsize=(10, 3))
izq.plot(x, np.cos(x))
izq.set_title("coseno")
der.plot(x, np.sin(x), color="red", label="seno")
der.plot(x, np.cos(x), color="blue", label="coseno")
der.legend()
plt.show()
```

## 5. Función con una discontinuidad

Dibujar $f(x) = \frac{1}{x+1}$ en $[-2, 2]$ con 100 puntos azules, evitando $x = -1$, donde se dibuja un punto rojo en $(-1, 0)$.

```python
x = np.linspace(-2, 2, 100)
x = x[x != -1]
plt.plot(x, 1 / (x + 1), "b.")
plt.plot(-1, 0, "ro")
plt.ylim(-20, 20)
plt.show()
```

## 6. Parábola con sus cortes

Dados $a$, $b$ y $c$, dibujar $f(x) = ax^2 + bx + c$ en $[z_1, z_2]$ ajustando los límites del gráfico al mínimo y máximo de $f$; dibujar los ejes en azul y con círculos rojos los puntos de corte con el eje horizontal.

```python
a, b, c = (float(input(f"{n}: ")) for n in "abc")
z1, z2 = float(input("z1: ")), float(input("z2: "))
x = np.linspace(z1, z2, 300)
y = a * x**2 + b * x + c
plt.plot(x, y)
plt.ylim(y.min() - 1, y.max() + 1)
plt.axhline(0, color="blue")
plt.axvline(0, color="blue")
d = b**2 - 4 * a * c
if a != 0 and d >= 0:
    raices = [(-b + s * d**0.5) / (2 * a) for s in (1, -1)]
    plt.plot(raices, [0, 0], "o", markersize=12, markerfacecolor="none", color="red")
plt.show()
```
