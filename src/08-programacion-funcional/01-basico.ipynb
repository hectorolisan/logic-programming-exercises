## 1. Descuento o IVA a una cesta

Escribir una función que aplique un descuento a un precio y otra que aplique el IVA. Escribir una tercera que reciba un diccionario `precio → porcentaje` y una de las anteriores, la aplique a cada producto y devuelva el precio final de la cesta.

```python
def aplica_descuento(precio, porcentaje):
    return precio * (1 - porcentaje / 100)

def aplica_iva(precio, porcentaje):
    return precio * (1 + porcentaje / 100)

def precio_cesta(cesta, funcion):
    return sum(funcion(precio, porcentaje) for precio, porcentaje in cesta.items())

cesta = {1000: 20, 500: 10, 100: 1}
print(precio_cesta(cesta, aplica_descuento), precio_cesta(cesta, aplica_iva))
```

## 2. Calculadora científica

Escribir una función que pregunte un valor y una función (seno, coseno, tangente, exponencial o logaritmo neperiano) y muestre una tabla con los enteros de 1 al valor y el resultado de aplicarles la función.

```python
import math

def calculadora():
    funciones = {"sin": math.sin, "cos": math.cos, "tan": math.tan, "exp": math.exp, "log": math.log}
    n = int(input("Valor: "))
    f = funciones[input(f"Función {list(funciones)}: ")]
    for i in range(1, n + 1):
        print(f"{i:3} {f(i):12.6f}")

calculadora()
```

## 3. `map` propio

Escribir una función que reciba otra función y una lista, y devuelva la lista con la función aplicada a cada elemento.

```python
def aplica(funcion, lista):
    return [funcion(x) for x in lista]

print(aplica(lambda x: x**2, [1, 2, 3, 4]))
```

## 4. `filter` propio

Escribir una función que reciba una función booleana y una lista, y devuelva los elementos para los que la función devuelve `True`.

```python
def filtra(funcion, lista):
    return [x for x in lista if funcion(x)]

print(filtra(lambda x: x % 2 == 0, range(10)))
```

## 5. Longitud de las palabras

Escribir una función que reciba una frase y devuelva un diccionario con cada palabra y su longitud.

```python
def longitudes(frase):
    return {palabra: len(palabra) for palabra in frase.split()}

print(longitudes("Programar en Python es divertido"))
```

## 6. Calificaciones

Escribir una función que reciba una lista de notas y devuelva sus calificaciones: SS (< 5), AP (< 7), NT (< 9), SB (< 10) y MH (10).

```python
def calificacion(nota):
    if nota < 5:
        return "SS"
    if nota < 7:
        return "AP"
    if nota < 9:
        return "NT"
    return "SB" if nota < 10 else "MH"

def calificaciones(notas):
    return list(map(calificacion, notas))

print(calificaciones([6.5, 5, 3.4, 8.2, 2.1, 9.7, 10]))
```

## 7. Asignaturas en mayúsculas

Escribir una función que reciba un diccionario `asignatura → nota` y devuelva otro con las asignaturas en mayúsculas y su calificación.

```python
def boletin(notas):
    return {asignatura.upper(): calificacion(nota) for asignatura, nota in notas.items()}

notas = {"Matemáticas": 6.5, "Física": 5, "Química": 3.4, "Economía": 8.2, "Historia": 9.7, "Programación": 10}
print(boletin(notas))
```

## 8. Solo aprobadas

Igual que el anterior, pero devolviendo solo las asignaturas aprobadas.

```python
def boletin_aprobadas(notas):
    return {asignatura.upper(): calificacion(nota) for asignatura, nota in notas.items() if nota >= 5}

print(boletin_aprobadas(notas))
```

## 9. Módulo de un vector

Escribir una función que calcule el módulo de un vector.

```python
from math import sqrt

def modulo(vector):
    return sqrt(sum(map(lambda x: x**2, vector)))

print(modulo([3, 4]), modulo([1, 2, 2]))
```

## 10. Comprensiones de listas

Con comprensiones: filtrar los negativos y el cero de `[-4, -3, -2, -1, 0, 2, 4, 6]`; aplanar `[[1, 2, 3], [4, 5, 6], [7, 8, 9]]`; y crear la lista de tuplas `(n, 1, n, n², n³, n⁴, n⁵)` para $n$ de 0 a 10.

```python
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]
print([n for n in numbers if n <= 0])
list_of_lists = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print([x for fila in list_of_lists for x in fila])
# n se repite delante de sus potencias 0..5
for t in [(n, *(n**e for e in range(6))) for n in range(11)]:
    print(t)
```

## 11. Transformar estructuras

A partir de `countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]`, obtener `[['FINLAND', 'FIN', 'HELSINKI'], ...]` y `[{'country': 'FINLAND', 'city': 'HELSINKI'}, ...]`. Con `names = [[('Asabeneh', 'Yetayeh')], [('David', 'Smith')], ...]`, obtener los nombres completos.

```python
countries = [[("Finland", "Helsinki")], [("Sweden", "Stockholm")], [("Norway", "Oslo")]]
print([[p.upper(), p[:3].upper(), c.upper()] for [(p, c)] in countries])
print([{"country": p.upper(), "city": c.upper()} for [(p, c)] in countries])
names = [[("Asabeneh", "Yetayeh")], [("David", "Smith")], [("Donald", "Trump")], [("Bill", "Gates")]]
print([f"{nombre} {apellido}" for [(nombre, apellido)] in names])
```

## 12. Lambdas para rectas

Escribir lambdas que calculen la pendiente y el corte con el eje $y$ de la recta que pasa por dos puntos.

```python
pendiente = lambda p, q: (q[1] - p[1]) / (q[0] - p[0])
corte_y = lambda p, q: p[1] - pendiente(p, q) * p[0]
print(pendiente((2, 2), (6, 10)), corte_y((2, 2), (6, 10)))
```
