# Diccionarios · Básico

Crear, consultar y recorrer diccionarios.

## 1. Símbolo de divisa

Guardar `{'Euro': '€', 'Dollar': '$', 'Yen': '¥'}`, pedir una divisa y mostrar su símbolo o un aviso si no está.

```python
divisas = {"Euro": "€", "Dollar": "$", "Yen": "¥"}
divisa = input("Divisa: ").title()
print(divisas.get(divisa, "Divisa no encontrada"))
```

## 2. Ficha personal

Pedir nombre, edad, dirección y teléfono, guardarlos en un diccionario y mostrar `<nombre> tiene <edad> años, vive en <dirección> y su número de teléfono es <teléfono>`.

```python
persona = {campo: input(f"{campo.capitalize()}: ") for campo in ("nombre", "edad", "dirección", "teléfono")}
print("{nombre} tiene {edad} años, vive en {dirección} y su número de teléfono es {teléfono}".format(**persona))
```

## 3. Frutería

Guardar los precios por kilo (plátano 1.35, manzana 0.80, pera 0.85, naranja 0.70), pedir fruta y kilos y mostrar el precio, o un aviso si la fruta no está.

```python
precios = {"plátano": 1.35, "manzana": 0.80, "pera": 0.85, "naranja": 0.70}
fruta = input("Fruta: ").lower()
kilos = float(input("Kilos: "))
if fruta in precios:
    print(f"{kilos} kg de {fruta} cuestan {kilos * precios[fruta]:.2f} €")
else:
    print("No tenemos", fruta)
```

## 4. Fecha con el nombre del mes

Pedir una fecha `dd/mm/aaaa` y mostrarla como `dd de <mes> de aaaa`.

```python
meses = {1: "enero", 2: "febrero", 3: "marzo", 4: "abril", 5: "mayo", 6: "junio", 7: "julio",
         8: "agosto", 9: "septiembre", 10: "octubre", 11: "noviembre", 12: "diciembre"}
dia, mes, año = input("Fecha (dd/mm/aaaa): ").split("/")
print(f"{dia} de {meses[int(mes)]} de {año}")
```

## 5. Créditos del curso

Guardar `{'Matemáticas': 6, 'Física': 4, 'Química': 5}`, mostrar `<asignatura> tiene <créditos> créditos` para cada una y el total del curso.

```python
creditos = {"Matemáticas": 6, "Física": 4, "Química": 5}
for asignatura, c in creditos.items():
    print(f"{asignatura} tiene {c} créditos")
print("Total:", sum(creditos.values()))
```

## 6. Rellenar un diccionario

Crear un diccionario vacío y rellenarlo con datos de una persona que se piden al usuario, mostrando el diccionario tras cada dato.

```python
persona = {}
for campo in ("nombre", "edad", "correo"):
    persona[campo] = input(f"{campo}: ")
    print(persona)
```
