# Diccionarios · Intermedio

Diccionarios como pequeñas bases de datos, contadores y traductores.

## 1. Lista de la compra

Pedir artículos y precios hasta que el usuario decida terminar, guardarlos en un diccionario y mostrar la lista con el coste total.

```python
cesta = {}
seguir = True
while seguir:
    articulo = input("Artículo: ")
    cesta[articulo] = float(input("Precio: "))
    seguir = input("¿Añadir otro? (s/n): ").lower() == "s"

print("Lista de la compra")
for articulo, precio in cesta.items():
    print(f"{articulo:<12}{precio:>8.2f}")
print(f"{'Total':<12}{sum(cesta.values()):>8.2f}")
```

## 2. Traductor español-inglés

El usuario introduce pares `palabra:traducción` separados por comas. Crear el diccionario, pedir una frase en español y traducirla palabra a palabra, dejando sin traducir las que no estén.

```python
pares = input("Pares palabra:traducción: ")
traductor = dict(par.strip().split(":") for par in pares.split(","))
frase = input("Frase: ")
print(" ".join(traductor.get(p, p) for p in frase.split()))
```

## 3. Facturas pendientes

Las facturas se guardan en un diccionario `número → coste`. En bucle, el usuario puede añadir una factura, pagarla (se elimina) o terminar. Tras cada operación, mostrar lo cobrado y lo pendiente.

```python
facturas, cobrado = {}, 0
while (opcion := input("¿(a)ñadir, (p)agar o (t)erminar? ")) != "t":
    numero = input("Número de factura: ")
    if opcion == "a":
        facturas[numero] = float(input("Coste: "))
    elif opcion == "p":
        cobrado += facturas.pop(numero, 0)
    print(f"Cobrado: {cobrado}  Pendiente: {sum(facturas.values())}")
```

## 4. Base de datos de clientes

Los clientes se guardan en un diccionario `NIF → datos` (nombre, dirección, teléfono, correo, preferente). Menú: (1) añadir, (2) eliminar, (3) mostrar, (4) listar todos, (5) listar preferentes, (6) terminar.

```python
clientes = {}
while (opcion := input("Opción (1-6): ")) != "6":
    if opcion == "1":
        nif = input("NIF: ")
        clientes[nif] = {
            "nombre": input("Nombre: "),
            "dirección": input("Dirección: "),
            "teléfono": input("Teléfono: "),
            "correo": input("Correo: "),
            "preferente": input("¿Preferente? (s/n): ") == "s",
        }
    elif opcion == "2":
        clientes.pop(input("NIF: "), None)
    elif opcion == "3":
        print(clientes.get(input("NIF: "), "No existe"))
    elif opcion in ("4", "5"):
        for nif, datos in clientes.items():
            if opcion == "4" or datos["preferente"]:
                print(nif, datos["nombre"])
```

## 5. Directorio de clientes desde texto

A partir de la cadena siguiente (primera línea = nombres de campos), crear un diccionario `nif → {campo: valor}` con el descuento como `float`.

```python
directorio = """nif;nombre;email;teléfono;descuento
01234567L;Luis González;luisgonzalez@mail.com;656343576;12.5
71476342J;Macarena Ramírez;macarena@mail.com;692839321;8
63823376M;Juan José Martínez;juanjo@mail.com;664888233;5.2"""

cabecera, *filas = [linea.split(";") for linea in directorio.splitlines()]
clientes = {}
for nif, *valores in filas:
    datos = dict(zip(cabecera[1:], valores))
    datos["descuento"] = float(datos["descuento"])
    clientes[nif] = datos
print(clientes)
```

## 6. Moda

Calcular la moda (el elemento más repetido) de una lista de enteros contando apariciones con un diccionario.

```python
def moda(valores):
    cuenta = {}
    for v in valores:
        cuenta[v] = cuenta.get(v, 0) + 1
    return max(cuenta, key=cuenta.get)

print(moda([3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]))
```
