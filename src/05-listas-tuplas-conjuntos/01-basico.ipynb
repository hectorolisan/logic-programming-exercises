# Listas y tuplas · Básico

Crear, recorrer, ordenar y filtrar listas.

## 1. Asignaturas

Guardar las asignaturas de un curso (Matemáticas, Física, Química, Historia y Lengua) en una lista y mostrarla.

```python
asignaturas = ["Matemáticas", "Física", "Química", "Historia", "Lengua"]
print(asignaturas)
```

## 2. Yo estudio...

Con la lista anterior, mostrar `Yo estudio <asignatura>` para cada asignatura.

```python
for asignatura in asignaturas:
    print("Yo estudio", asignatura)
```

## 3. Notas por asignatura

Pedir la nota de cada asignatura y después mostrar `En <asignatura> has sacado <nota>`.

```python
notas = [float(input(f"Nota de {a}: ")) for a in asignaturas]
for asignatura, nota in zip(asignaturas, notas):
    print(f"En {asignatura} has sacado {nota}")
```

## 4. Lotería primitiva

Pedir los 6 números ganadores, guardarlos en una lista y mostrarlos ordenados de menor a mayor.

```python
numeros = [int(input(f"Número {i}: ")) for i in range(1, 7)]
print(sorted(numeros))
```

## 5. Del 10 al 1

Guardar en una lista los números del 1 al 10 y mostrarlos en orden inverso separados por comas.

```python
numeros = list(range(1, 11))
print(", ".join(map(str, reversed(numeros))))
```

## 6. Abecedario sin múltiplos de 3

Guardar el abecedario en una lista, eliminar las letras en posiciones múltiplo de 3 (contando desde 1) y mostrar el resultado.

```python
from string import ascii_lowercase

letras = list(ascii_lowercase)
print([l for i, l in enumerate(letras, 1) if i % 3])
```

## 7. Palíndromo

Pedir una palabra y decir si es un palíndromo.

```python
palabra = input("Palabra: ").lower()
print("Es palíndromo" if palabra == palabra[::-1] else "No es palíndromo")
```

## 8. Precio menor y mayor

Guardar los precios 50, 75, 46, 22, 80, 65 y 8 en una lista y mostrar el menor y el mayor.

```python
precios = [50, 75, 46, 22, 80, 65, 8]
print("Menor:", min(precios), "Mayor:", max(precios))
```

## 9. Negativos a cero

Dada una lista cualquiera, sustituir los elementos negativos por cero (modificando la propia lista).

```python
a = [3, -1, 0, -7, 5]
for i, v in enumerate(a):
    if v < 0:
        a[i] = 0
print(a)
```

## 10. Eliminar índices pares

Eliminar de una lista los elementos de índice par. Ejemplo: `[1, 2, 1, 5, 0, 3]` → `[2, 5, 3]`.

```python
a = [1, 2, 1, 5, 0, 3]
del a[::2]
print(a)
```

## 11. Eliminar valores pares

Eliminar de una lista los elementos de valor par. Ejemplo: `[1, -2, 1, -5, 0, 3]` → `[1, 1, -5, 3]`.

```python
a = [1, -2, 1, -5, 0, 3]
a[:] = [v for v in a if v % 2]
print(a)
```
