Listas con cálculos: estadística, vectores, primos y barajas.

## 1. Asignaturas a repetir

Pedir la nota de cada asignatura, eliminar de la lista las aprobadas y mostrar las que hay que repetir.

```python
asignaturas = ["Matemáticas", "Física", "Química", "Historia", "Lengua"]
notas = [float(input(f"Nota de {a}: ")) for a in asignaturas]
suspensas = [a for a, n in zip(asignaturas, notas) if n < 5]
print("Tienes que repetir:", suspensas)
```

## 2. Vocales de una palabra

Pedir una palabra y mostrar cuántas veces contiene cada vocal.

```python
palabra = input("Palabra: ").lower()
for vocal in "aeiou":
    print(vocal, palabra.count(vocal))
```

## 3. Producto escalar

Guardar los vectores (1, 2, 3) y (-1, 0, 2) en dos listas y mostrar su producto escalar.

```python
u, v = [1, 2, 3], [-1, 0, 2]
print(sum(a * b for a, b in zip(u, v)))
```

## 4. Media y desviación típica

Pedir una muestra de números separados por comas, guardarlos en una lista y mostrar la media y la desviación típica.

```python
muestra = [float(x) for x in input("Muestra: ").split(",")]
media = sum(muestra) / len(muestra)
desviacion = (sum((x - media) ** 2 for x in muestra) / len(muestra)) ** 0.5
print(f"Media: {media}  Desviación típica: {desviacion}")
```

## 5. Los n primeros primos

Construir una lista con los $n$ primeros números primos (no los primos menores que $n$).

```python
n = int(input("n: "))
primos, candidato = [], 2
while len(primos) < n:
    # basta dividir por primos previos hasta la raíz
    if all(candidato % p for p in primos if p * p <= candidato):
        primos.append(candidato)
    candidato += 1
print(primos)
```

## 6. Leer 10 positivos

Leer una lista de 10 enteros asegurando que todos son positivos: si alguno es negativo, avisar y volver a pedirlo.

```python
numeros = []
while len(numeros) < 10:
    n = int(input(f"Número {len(numeros) + 1}: "))
    if n < 0:
        print("Debe ser positivo")
    else:
        numeros.append(n)
print(numeros)
```

## 7. Producto cartesiano sin `itertools`

Escribir `producto`, que se comporte como `itertools.product` con dos iterables, y usarla para generar la baraja de póker (A, 2-10, J, Q, K × picas, corazones, tréboles y diamantes) y la baraja española (1-10 × oros, copas, espadas y bastos).

```python
import itertools

def producto(a, b):
    return [(x, y) for x in a for y in b]

numeros_poker = ["A", *map(str, range(2, 11)), "J", "Q", "K"]
palos_poker = ["Picas", "Corazones", "Tréboles", "Diamantes"]
poker = producto(numeros_poker, palos_poker)
assert poker == list(itertools.product(numeros_poker, palos_poker))
print(len(poker), poker[:5])

española = producto(range(1, 11), ["Oros", "Copas", "Espadas", "Bastos"])
print(len(española), española[:5])
```

## 8. ¿Es aleatorio `shuffle`?

Barajar cientos de veces la baraja de póker ordenada con `random.shuffle` y medir con qué frecuencia una carta acaba en cada posición. Si es aleatorio, cada posición debería rondar $1/52 \approx 0.019$.

```python
import random
from collections import Counter

def baraja_cartas(cartas):
    copia = cartas[:]
    random.shuffle(copia)
    return copia

simulaciones = 20000
posiciones = Counter(baraja_cartas(poker).index(("A", "Picas")) for _ in range(simulaciones))
frecuencias = [posiciones[i] / simulaciones for i in range(52)]
print(f"mín {min(frecuencias):.4f}  máx {max(frecuencias):.4f}  esperado {1 / 52:.4f}")
```
