## 1. Indexar listas

¿Qué dan `[1][0]` y `[][0]`?

## 2. Copias de listas

¿Sirve `y = x + []` para copiar una lista igual que `y = x[:]`? Sí: ambas crean una lista nueva.

```python
x = [1, 2, 3]
y1, y2, alias = x[:], x + [], x
x[0] = 99
print(y1, y2, alias, y1 is x, y2 is x)
```

## 4. Comparar listas sin operadores de comparación

Decir si una lista es menor que otra (orden lexicográfico) sin usar `<` entre listas.

```python
def menor(a, b):
    for x, y in zip(a, b):
        if x != y:
            return x < y
    return len(a) < len(b)

for a, b in (([1, 2], [1, 3]), ([1, 2], [1, 2, 0]), ([2], [1, 9]), ([1, 2], [1, 2])):
    assert menor(a, b) == (a < b)
    print(a, b, menor(a, b))
```

## 5. Identidad e igualdad

Explicar `a is a`, `a + [] is a`, `a + [] == a`, y los resultados de comparar con `==` e `is` listas iguales, sus elementos y sus cortes.

`==` compara contenido; `is`, si son el mismo objeto. Los enteros pequeños se reutilizan, por eso `a[0] is b[0]` puede ser `True`.

```python
a = [1, 2, 3]
print(a is a, a + [] is a, a + [] == a)
a, b = [1, 2, 1], [1, 2, 1]
print(all(x is y for x, y in zip(a, b)), a == b, a is b)
print([1, 2] == [1, 2], [1, 2] is [1, 2])
a = [1, 2, 3]
b = [a[0], a[1], a[2]]
print(a == b, a is b, a[0] == b[1], b is [b[0], b[1], b[2]])
a = [1, 2, 3, 4, 5]
b, c, d = a[1:3], a, a[:]
print(a == c, a == d, c == d, a == b, a is c, a is d, c is d, a is b)
```

## 7. Alias y copias en memoria

¿Qué muestra este programa? `c = a` es un alias de `a`; `d = b[:]` y `e = a + b` son listas nuevas.

```python
a = list(range(5))
b = list(range(5))
c = a
d = b[:]
e = a + b
f = b[:1]
g = b[0]
c[0] = 100
d[0] = 200
e[0] = 300
print(a, b, c, d, e, f, g)

a = [1, 2, 1]
b = 1
c = [2, 1, 2]
d = c
d[2] = 3
e = d[:1]
f = d[:]
f[0] = a[1]
f[1] = 1
print(a, b, c, d, e, f)
```

## 10. Pertenencia con error

¿Por qué falla este programa? El `break` en el `else` sale del bucle en cuanto el primer elemento no coincide.

```python
elemento, lista = 5, [1, 4, 5, 1, 3, 8]
for i in lista:
    if elemento == i:
        pertenece = True
    else:
        pertenece = False
        break
print("Versión errónea:", pertenece)
pertenece = False
for i in lista:
    if elemento == i:
        pertenece = True
        break
print("Corregida:", pertenece, elemento in lista)
```

## 11. Letras minúsculas con `in`

¿Qué hacen dos programas que dicen si una letra es minúscula comprobando `'a' <= letra <= 'z'` o la pertenencia a las letras con tilde, diéresis y eñe? Ambos aciertan con letras sueltas. La diferencia está en la longitud: `letra in 'áéíóúüñ'` sin comprobar `len(letra) == 1` daría por buenas subcadenas como `'éí'` o la cadena vacía.

```python
def minuscula_1(letra):
    return (len(letra) == 1 and "a" <= letra <= "z") or letra in ["á", "é", "í", "ó", "ú", "ü", "ñ"]

def minuscula_2(letra):
    return len(letra) == 1 and ("a" <= letra <= "z" or letra in "áéíóúüñ")

for l in ("a", "ñ", "A", "éí", ""):
    print(repr(l), minuscula_1(l), minuscula_2(l), "(sin len)", l in "áéíóúüñ")
```

## 12. Burbuja con cadenas

¿Qué ocurre si el método de la burbuja ordena `['Pepe', 'Juan', 'María', 'Ana', 'Luis', 'Pedro']`? Funciona igual: las cadenas se comparan alfabéticamente.

```python
lista = ["Pepe", "Juan", "María", "Ana", "Luis", "Pedro"]
for i in range(len(lista)):
    for j in range(len(lista) - 1 - i):
        if lista[j] > lista[j + 1]:
            lista[j], lista[j + 1] = lista[j + 1], lista[j]
print(lista)
```

## 14. Matrices nula e identidad

Construir una matriz nula de 5 × 5 y una identidad de 4 × 4 con bucles (cuidado: `[[0] * 5] * 5` repite la misma fila).

```python
nula = [[0] * 5 for _ in range(5)]
identidad = []
for i in range(4):
    fila = [0] * 4
    fila[i] = 1
    identidad.append(fila)
print(nula, identidad)
mal = [[0] * 3] * 3
mal[0][0] = 1
print("Filas compartidas:", mal)
```

## 15. Recorrer matrices

¿Qué muestran estos accesos a la identidad 3 × 3 y la media de sus elementos?

```python
M = [[1, 0, 0], [0, 1, 0], [0, 0, 1]]
print(M[-1][0], M[-1][-1])
for fila in M:
    print(fila)
print([M[i][j] for i in range(3) for j in range(3)])
print(sum(sum(fila) for fila in M) / 9.0)
```

## 16. Escalar, traspuesta y triangular superior

Multiplicar una matriz por un número, obtener la traspuesta y decir si es diagonal (triangular) superior.

```python
A = [[1, 2, 3], [2, 12, 6], [1, 0, -3], [10, -1, 0]]
print([[3 * x for x in fila] for fila in A])
print([list(col) for col in zip(*A)])

def diagonal_superior(M):
    return all(M[i][j] == 0 for i in range(len(M)) for j in range(min(i, len(M[0]))))

print(diagonal_superior([[1, 2, 3], [0, 12, 6], [0, 0, 0]]), diagonal_superior(A))
```
