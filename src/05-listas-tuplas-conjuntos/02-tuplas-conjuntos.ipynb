## 2. Alimentos

Crear tuplas de frutas, verduras y productos animales y unirlas en `food_stuff_tp`; convertirla a lista; recortar los elementos centrales y los 3 primeros y 3 últimos; y borrar la tupla.

```python
frutas = ("plátano", "naranja", "mango")
verduras = ("tomate", "papa", "repollo")
animales = ("leche", "carne", "huevo", "queso")
food_stuff_tp = frutas + verduras + animales
food_stuff_lt = list(food_stuff_tp)
medio = len(food_stuff_lt) // 2
print(food_stuff_lt[medio - 1:medio + 1] if len(food_stuff_lt) % 2 == 0 else food_stuff_lt[medio])
print(food_stuff_lt[:3], food_stuff_lt[-3:])
del food_stuff_tp
```

## 5. Operaciones con conjuntos

Con `it_companies` (conjunto), $A = \{19, 22, 24, 20, 25, 26\}$ y $B = \{19, 22, 20, 25, 26, 24, 28, 27\}$: añadir una y varias empresas, quitar una (diferencia entre `remove` y `discard`), y calcular unión, intersección, si $A \subseteq B$, si son disjuntos y la diferencia simétrica.

```python
it_companies = {"Facebook", "Google", "Microsoft", "Apple", "IBM", "Oracle", "Amazon"}
print(len(it_companies))
it_companies.add("Twitter")
it_companies.update({"Nvidia", "Intel"})
it_companies.remove("IBM")
# discard no falla si no existe; remove lanza KeyError
it_companies.discard("No existe")
print(sorted(it_companies))

A = {19, 22, 24, 20, 25, 26}
B = {19, 22, 20, 25, 26, 24, 28, 27}
print(A | B, A & B, A <= B, A.isdisjoint(B), A ^ B)
del A, B
```
