# Tipos de datos simples · Básico

Variables, `print`, `input`, conversión de tipos y operadores aritméticos.

## 1. Hola Mundo

Mostrar por pantalla la cadena `¡Hola Mundo!`.

```python
print("¡Hola Mundo!")
```

## 2. Hola Mundo en una variable

Guardar `¡Hola Mundo!` en una variable y mostrar su contenido.

```python
mensaje = "¡Hola Mundo!"
print(mensaje)
```

## 3. Saludo personalizado

Pedir el nombre del usuario y mostrar `¡Hola <nombre>!`.

```python
nombre = input("¿Cómo te llamas? ")
print(f"¡Hola {nombre}!")
```

## 4. Operación aritmética

Mostrar el resultado de $\left(\frac{3+2}{2\cdot5}\right)^2$.

```python
print(((3 + 2) / (2 * 5)) ** 2)
```

## 5. Paga semanal

Pedir horas trabajadas y coste por hora, y mostrar la paga.

```python
horas = float(input("Horas trabajadas: "))
coste = float(input("Coste por hora: "))
print("Paga:", horas * coste)
```

## 6. Suma de los n primeros enteros

Leer un entero positivo $n$ y mostrar la suma de 1 a $n$ usando $\frac{n(n+1)}{2}$.

```python
n = int(input("Entero positivo: "))
print(f"Suma de 1 a {n}: {n * (n + 1) // 2}")
```

## 7. Índice de masa corporal

Pedir peso (kg) y estatura (m) y mostrar `Tu índice de masa corporal es <imc>` con dos decimales.

```python
peso = float(input("Peso (kg): "))
estatura = float(input("Estatura (m): "))
imc = peso / estatura**2
print(f"Tu índice de masa corporal es {imc:.2f}")
```

## 8. Cociente y resto

Pedir dos enteros y mostrar `<n> entre <m> da un cociente <c> y un resto <r>`.

```python
n = int(input("Dividendo: "))
m = int(input("Divisor: "))
c, r = divmod(n, m)
print(f"{n} entre {m} da un cociente {c} y un resto {r}")
```
