## 5. Paga semanal

Pedir horas trabajadas y coste por hora, y mostrar la paga.

```python
horas = float(input("Horas trabajadas: "))
coste = float(input("Coste por hora: "))
print("Paga:", horas * coste)
```

## 6. Suma de los n primeros enteros

Leer un entero positivo $n$ y mostrar la suma de 1 a $n$ usando $\frac{n(n+1)}{2}$.

```python
n = int(input("Entero positivo: "))
print(f"Suma de 1 a {n}: {n * (n + 1) // 2}")
```

## 7. Índice de masa corporal

Pedir peso (kg) y estatura (m) y mostrar `Tu índice de masa corporal es <imc>` con dos decimales.

```python
peso = float(input("Peso (kg): "))
estatura = float(input("Estatura (m): "))
imc = peso / estatura**2
print(f"Tu índice de masa corporal es {imc:.2f}")
```

## 8. Cociente y resto

Pedir dos enteros y mostrar `<n> entre <m> da un cociente <c> y un resto <r>`.

```python
n = int(input("Dividendo: "))
m = int(input("Divisor: "))
c, r = divmod(n, m)
print(f"{n} entre {m} da un cociente {c} y un resto {r}")
```

## 9. Operadores aritméticos

Con `num_one = 5` y `num_two = 4`, calcular suma, resta, producto, división, resto de `num_two` entre `num_one`, potencia y división entera.

```python
num_one, num_two = 5, 4
total = num_one + num_two
diff = num_one - num_two
product = num_one * num_two
division = num_one / num_two
remainder = num_two % num_one
exp = num_one**num_two
floor_division = num_one // num_two
print(total, diff, product, division, remainder, exp, floor_division)
```

## 10. Comparaciones y conversiones

Evaluar: ¿miden distinto `'python'` y `'dragon'`?, ¿contienen ambas `'on'`?, ¿está `'jargon'` en *I hope this course is not full of jargon*?, la longitud de `'python'` como `float` y como `str`, ¿`7 // 3` es igual a `int(2.7)`?, ¿es igual el tipo de `'10'` que el de `10`? y ¿qué pasa con `int('9.8')`?

```python
print(len("python") != len("dragon"))
print("on" in "python" and "on" in "dragon")
print("jargon" in "I hope this course is not full of jargon")
print(float(len("python")), str(float(len("python"))))
print(7 // 3 == int(2.7))
print(type("10") == type(10))
try:
    print(int("9.8") == 10)
except ValueError as e:
    # int() no acepta cadenas con decimales
    print("Error:", e)
```
