# Tipos de datos simples · Intermedio

Fórmulas, redondeo, módulo `math` y división entera.

## 1. Inversión con interés compuesto

Pedir cantidad a invertir, interés anual (%) y número de años, y mostrar el capital final.

```python
cantidad = float(input("Cantidad a invertir: "))
interes = float(input("Interés anual (%): "))
años = int(input("Años: "))
print(f"Capital final: {cantidad * (1 + interes / 100) ** años:.2f}")
```

## 2. Peso de un paquete de juguetes

Cada payaso pesa 112 g y cada muñeca 75 g. Leer cuántos de cada se venden y mostrar el peso total del paquete.

```python
payasos = int(input("Payasos: "))
muñecas = int(input("Muñecas: "))
print(f"Peso total: {payasos * 112 + muñecas * 75} g")
```

## 3. Cuenta de ahorros al 4 %

Leer el depósito inicial y mostrar el saldo tras el 1.º, 2.º y 3.º año, redondeado a dos decimales.

```python
saldo = float(input("Depósito inicial: "))
for año in range(1, 4):
    saldo *= 1.04
    print(f"Año {año}: {saldo:.2f}")
```

## 4. Pan del día anterior

La barra cuesta 3.49 € y la que no es del día tiene un 60 % de descuento. Leer cuántas barras no frescas se venden y mostrar precio habitual, descuento y coste final.

```python
barras = int(input("Barras no frescas: "))
precio, descuento = 3.49, 0.6
print(f"Precio habitual: {precio} €")
print(f"Descuento: {descuento:.0%}")
print(f"Coste final: {barras * precio * (1 - descuento):.2f} €")
```

## 5. Área de un triángulo por sus tres lados

Pedir los tres lados y mostrar perímetro y área con la fórmula de Herón: $A=\sqrt{s(s-a)(s-b)(s-c)}$, con $s=\frac{a+b+c}{2}$.

Prueba: lados 3, 5 y 7 → perímetro 15.0, área ≈ 6.4952.

```python
from math import sqrt

a = float(input("Lado a: "))
b = float(input("Lado b: "))
c = float(input("Lado c: "))
s = (a + b + c) / 2
print("Perímetro:", a + b + c)
print("Área:", sqrt(s * (s - a) * (s - b) * (s - c)))
```

## 6. Área de un triángulo con dos lados y un ángulo

Pedir dos lados y el ángulo que forman (en grados) y mostrar el área $A=\frac{1}{2}ab\sin\alpha$.

Prueba: $a=1$, $b=2$, $\alpha=30$ → 0.5.

```python
from math import radians, sin

a = float(input("Lado a: "))
b = float(input("Lado b: "))
angulo = float(input("Ángulo (grados): "))
# sin trabaja en radianes
print(f"Área: {a * b * sin(radians(angulo)) / 2:.4f}")
```

## 7. Círculo

Pedir el radio y mostrar área y perímetro con 2 decimales.

```python
from math import pi

r = float(input("Radio: "))
print(f"Área: {pi * r**2:.2f}")
print(f"Perímetro: {2 * pi * r:.2f}")
```

## 8. Desglose en billetes y monedas

Desglosar una cantidad entera de euros en billetes de 500, 200, 100, 50, 20, 10 y 5 € y monedas de 2 y 1 €, mostrando solo los que aparecen.

Ejemplo: 434 → 2 billetes de 200, 1 de 20, 1 de 10 y 2 monedas de 2.

```python
cantidad = int(input("Cantidad (€): "))
for valor in (500, 200, 100, 50, 20, 10, 5, 2, 1):
    n, cantidad = divmod(cantidad, valor)
    if n:
        tipo = "billete" if valor >= 5 else "moneda"
        print(f"{n} {tipo}{'s' if n > 1 else ''} de {valor} euros")
```
