## 1. Ecuación de primer grado mal planteada

¿Es correcto `a*x+b=0` para resolver la ecuación? ¿Y calcular `x = -b / a` antes de leer `a` y `b`?

No: `a*x+b=0` no es una asignación válida (a la izquierda de `=` solo puede ir una variable) y en el segundo programa `a` y `b` aún no existen al calcular `x`.

```python
for codigo in ("a*x+b=0", "x = -b / a\na = 2.0\nb = 4.0"):
    try:
        exec(compile(codigo, "<programa>", "exec"), {})
    except (SyntaxError, NameError) as e:
        print(type(e).__name__, "->", e)
```

## 3. Mensajes de más

Si se quita la segunda condición y se deja el `print` sin `if`, ¿qué sale con `a = 2` y `b = 2`? Trazar el programa para `a=2, b=6`, `a=0, b=3`, `a=0, b=-3` y `a=0, b=0`.

Con `a = 2` se muestra la solución y, además, el mensaje de "no tiene solución", que es falso.

```python
def primer_grado_mal(a, b):
    lineas = [4]
    if a != 0:
        lineas += [5, 6]
        print("Solución:", -b / a)
    lineas.append(8)
    print("La ecuación no tiene solución.")
    return lineas

for a, b in ((2, 2), (2, 6), (0, 3), (0, -3), (0, 0)):
    print(f"a={a}, b={b}: líneas {primer_grado_mal(a, b)}")
```

## 5. ¿Qué entradas producen cada mensaje?

Analizar dos programas: uno clasifica una minúscula según sea `<= 'k'` o `>= 'l'`, y otro calcula el cuadrante de un ángulo con `int(ceil(grados) % 360) // 90`. ¿Qué muestra un programa que compara `14 < 120` y `'14' < '120'`?

```python
from math import ceil

for letra in "akl z":
    print(repr(letra), "primeras" if letra <= "k" else "", "últimas" if letra >= "l" else "")

def cuadrante(grados):
    return ["primer", "segundo", "tercer", "cuarto"][int(ceil(grados) % 360) // 90] + " cuadrante"

for g in (0, 45, 90, 180.5, 359, 360, -10):
    print(g, cuadrante(g))

if 14 < 120:
    print("Primer saludo")
# '1' == '1' y '4' > '2': la cadena '14' es mayor
if "14" < "120":
    print("Segundo saludo")
```

## 8. Años para alcanzar un capital

Calcular cuántos años se tarda en convertir $C_i$ en $C_f$ al $x$ %: $n = \frac{\log C_f - \log C_i}{\log(1 + x/100)}$, controlando divisiones por cero y logaritmos inválidos (si $C_i = C_f$, son 0 años).

Ejemplos: 10 000 → 11 000 al 5 % = 1.9535 años; al 0 % es imposible.

```python
from math import log

def años_necesarios(inicial, final, x):
    if inicial == final:
        return 0
    if inicial <= 0 or final <= 0 or x <= -100 or x == 0 or (final > inicial) != (x > 0):
        return None
    return (log(final) - log(inicial)) / log(1 + x / 100)

print(round(años_necesarios(10000, 11000, 5), 4), años_necesarios(10000, 11000, 0), años_necesarios(10000, 10000, 0))
```

## 9. Versiones de la ecuación de segundo grado

¿Hay diferencias al anidar `if` frente a usar condiciones con `and`? ¿Qué pasa al comprobar `sqrt(discriminante) >= 0` con `a = 4, b = 2, c = 4`?

Las dos estructuras son equivalentes si cubren los mismos casos. El último programa falla antes de comparar: `sqrt` de un número negativo lanza `ValueError`.

```python
from math import sqrt

def version_anidada(a, b, c):
    if a == 0:
        if b == 0:
            return "infinitas" if c == 0 else "sin solución"
        return -c / b
    d = b**2 - 4 * a * c
    return "sin solución real" if d < 0 else ((-b + sqrt(d)) / (2 * a), (-b - sqrt(d)) / (2 * a))

def version_and(a, b, c):
    if a == 0 and b == 0 and c == 0:
        return "infinitas"
    if a == 0 and b == 0:
        return "sin solución"
    if a == 0:
        return -c / b
    d = b**2 - 4 * a * c
    return "sin solución real" if d < 0 else ((-b + sqrt(d)) / (2 * a), (-b - sqrt(d)) / (2 * a))

for coef in ((0, 0, 0), (0, 0, 1), (0, 2, 4), (1, -3, 2), (4, 2, 4)):
    assert version_anidada(*coef) == version_and(*coef)
try:
    sqrt(2**2 - 4 * 4 * 4)
except ValueError as e:
    print("ValueError:", e)
```

## 12. Máximo de dos y tres números

¿Qué líneas se ejecutan en el programa del máximo para `(2, 3)`, `(3, 2)`, `(−2, 0)` y `(1, 1)`? ¿Es correcto un programa con `if a > b` y `if b > a` sin `else`? ¿Qué pasa con tres números iguales?

Con dos números iguales ninguna condición se cumple y `maximo` queda sin definir (`NameError`).

```python
def maximo_traza(a, b):
    lineas = []
    if a > b:
        lineas.append(5)
        m = a
    else:
        lineas.append(7)
        m = b
    return m, lineas

for a, b in ((2, 3), (3, 2), (-2, 0), (1, 1)):
    print(a, b, maximo_traza(a, b))

def maximo_aprendiz(a, b):
    if a > b:
        maximo = a
    if b > a:
        maximo = b
    return maximo

try:
    maximo_aprendiz(4, 4)
except UnboundLocalError as e:
    print("Error:", e)

def maximo3(a, b, c):
    m = a
    if b > m:
        m = b
    if c > m:
        m = c
    return m

print(maximo3(2, 3, 4), maximo3(3, 2, 4), maximo3(1, 1, 1))
```

## 13. Menor de cinco palabras

Calcular la primera de cinco palabras en orden alfabético: primero aceptando que las mayúsculas van antes (orden ASCII); después ignorando mayúsculas (`'pepita'` < `'Pepito'`).

```python
palabras = [input("Palabra: ") for _ in range(5)]
menor = palabras[0]
for p in palabras[1:]:
    if p < menor:
        menor = p
print("Orden ASCII:", menor)

palabras = [input("Palabra: ") for _ in range(5)]
menor = palabras[0]
for p in palabras[1:]:
    if p.lower() < menor.lower():
        menor = p
print("Sin mayúsculas:", menor)
```

## 15. ¿Qué valores provocan cada mensaje?

Analizar cuatro programas (aparcamiento según el día, estaciones según el mes, carácter válido en un identificador y año bisiesto) probándolos con valores representativos.

```python
def aparcar(dia):
    if 0 < dia <= 15:
        return "lado izquierdo"
    return "lado derecho" if 15 < dia < 32 else f"Ningún mes tiene {dia} días."

def estacion(mes):
    if 1 <= mes <= 3:
        return "Invierno"
    if mes in (4, 5, 6):
        return "Primavera"
    if not (mes < 7 or 9 < mes):
        return "Verano"
    return "Otoño" if not (mes != 10 and mes != 11 and mes != 12) else f"Ningún año tiene {mes} meses."

def caracter_identificador(car):
    # ojo: el espacio no es válido en un identificador; el programa original se equivoca
    if "a" <= car.lower() <= "z" or car == " ":
        return "válido"
    return "dígito: válido salvo al principio" if not (car < "0" or "9" < car) else "no válido"

print([aparcar(d) for d in (0, 15, 16, 31, 32)])
print([estacion(m) for m in (2, 5, 8, 11, 13)])
print([caracter_identificador(c) for c in "aZ 5_"])
```

## 16. Evaluación en cortocircuito

¿Por qué `if 1/a > 1 and a != 0:` falla con `a = 0`? Porque `and` evalúa de izquierda a derecha: hay que poner primero `a != 0`.

```python
a = 0
try:
    if 1 / a > 1 and a != 0:
        print(a)
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)
if a != 0 and 1 / a > 1:
    print(a)
print("Sin error al invertir el orden")
```

Un aprendiz escribe `if opcion == a:` sin comillas. ¿Funciona? Corregirlo, hacer una traza con la opción `d`, aceptar mayúsculas y reescribirlo con `elif`.

## 18. Calificación con `elif`

Reescribir la calificación cualitativa (Suspenso, Aprobado, Notable, Sobresaliente, Matrícula de Honor) con `elif` y comprobarla.

## 19. Depuración paso a paso

Los ejercicios que piden ejecutar paso a paso con el depurador de PythonG se hacen hoy con el depurador del editor (VS Code o Jupyter): colocar un punto de ruptura y avanzar línea a línea. Como alternativa en código, `sys.settrace` permite registrar las líneas ejecutadas.

```python
import sys

def lineas_ejecutadas(funcion, *args):
    lineas = []
    def traza(frame, evento, arg):
        if evento == "line" and frame.f_code is funcion.__code__:
            lineas.append(frame.f_lineno - funcion.__code__.co_firstlineno)
        return traza
    sys.settrace(traza)
    try:
        funcion(*args)
    finally:
        sys.settrace(None)
    return lineas

print(lineas_ejecutadas(maximo3, 2, 3, 4), lineas_ejecutadas(maximo3, 1, 1, 1))
```
