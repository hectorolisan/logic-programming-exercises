## 1. Grupos A y B

El grupo A lo forman las mujeres con nombre anterior a la M y los hombres con nombre posterior a la N; el resto va al grupo B. Pedir nombre y sexo (M/H) y mostrar el grupo.

```python
nombre = input("Nombre: ").lower()
sexo = input("Sexo (M/H): ").upper()
grupo_a = (sexo == "M" and nombre < "m") or (sexo == "H" and nombre > "n")
print("Grupo", "A" if grupo_a else "B")
```

## 2. Tipo impositivo por tramos

Pedir la renta anual y mostrar el tipo impositivo: menos de 10000 € → 5 %, hasta 20000 € → 15 %, hasta 35000 € → 20 %, hasta 60000 € → 30 %, más de 60000 € → 45 %.

```python
renta = float(input("Renta anual: "))
if renta < 10000:
    tipo = 5
elif renta < 20000:
    tipo = 15
elif renta < 35000:
    tipo = 20
elif renta < 60000:
    tipo = 30
else:
    tipo = 45
print(f"Tipo impositivo: {tipo} %")
```

## 3. Evaluación de empleados

Las puntuaciones válidas son 0.0 (Inaceptable), 0.4 (Aceptable) o 0.6 o más (Meritorio); no hay valores intermedios. El premio es 2400 € × puntuación. Leer la puntuación y mostrar nivel y dinero, o un aviso si no es válida.

```python
puntos = float(input("Puntuación: "))
if puntos == 0.0:
    nivel = "Inaceptable"
elif puntos == 0.4:
    nivel = "Aceptable"
elif puntos >= 0.6:
    nivel = "Meritorio"
else:
    nivel = None
if nivel:
    print(f"Nivel {nivel}. Recibirás {2400 * puntos:.2f} €")
else:
    print("Puntuación no válida")
```

## 4. Precio de entrada

Menores de 4 años entran gratis, de 4 a 18 años pagan 5 € y mayores de 18, 10 €. Pedir la edad y mostrar el precio.

```python
edad = int(input("Edad: "))
if edad < 4:
    precio = 0
elif edad <= 18:
    precio = 5
else:
    precio = 10
print(f"Precio: {precio} €")
```

## 5. Pizzería Bella Napoli

Preguntar si quiere pizza vegetariana. Mostrar los ingredientes disponibles (vegetarianos: pimiento y tofu; no vegetarianos: peperoni, jamón y salmón), dejar elegir uno y mostrar el tipo de pizza con todos sus ingredientes (siempre llevan mozzarella y tomate).

```python
vegetariana = input("¿Pizza vegetariana? (s/n): ").lower() == "s"
opciones = ["Pimiento", "Tofu"] if vegetariana else ["Peperoni", "Jamón", "Salmón"]
ingrediente = input(f"Elige ingrediente {opciones}: ")
if ingrediente in opciones:
    print(f"Pizza {'vegetariana' if vegetariana else 'no vegetariana'} con mozzarella, tomate y {ingrediente.lower()}")
else:
    print("Ingrediente no disponible")
```

## 6. Máximo de cinco números

Calcular el máximo de 5 enteros usando un candidato a máximo que se actualiza al comparar con cada número (sin `max`).

```python
maximo = int(input("Número 1: "))
for i in range(2, 6):
    n = int(input(f"Número {i}: "))
    if n > maximo:
        maximo = n
print("Máximo:", maximo)
```

## 7. Punto más cercano

Dados cinco puntos del plano, decir cuál de los cuatro últimos está más cerca del primero.

```python
from math import dist

origen = [float(v) for v in input("Punto 1 (x y): ").split()]
mejor, mejor_d = None, float("inf")
for i in range(2, 6):
    p = [float(v) for v in input(f"Punto {i} (x y): ").split()]
    if dist(origen, p) < mejor_d:
        mejor, mejor_d = i, dist(origen, p)
print(f"El más cercano es el punto {mejor} (distancia {mejor_d:.3f})")
```

## 8. Calificación cualitativa

Dada una nota (0-10), mostrar Suspenso (< 5), Aprobado (< 7), Notable (< 8.5), Sobresaliente (< 10) o Matrícula de Honor (10).

```python
nota = float(input("Nota: "))
if not 0 <= nota <= 10:
    print("Nota no válida")
elif nota < 5:
    print("Suspenso")
elif nota < 7:
    print("Aprobado")
elif nota < 8.5:
    print("Notable")
elif nota < 10:
    print("Sobresaliente")
else:
    print("Matrícula de Honor")
```

## 9. Tipo de carácter

Clasificar un carácter como vocal minúscula, vocal mayúscula, consonante minúscula, consonante mayúscula u otro.

```python
c = input("Carácter: ")
if len(c) != 1 or not c.isalpha():
    print("Otro tipo de carácter")
else:
    tipo = "vocal" if c.lower() in "aeiouáéíóúü" else "consonante"
    print(tipo, "mayúscula" if c.isupper() else "minúscula")
```

## 10. Año bisiesto

Un año es bisiesto si es divisible por 4 y no por 100, salvo que también lo sea por 400. Pedir un año y mostrar cuántos días tiene.

```python
año = int(input("Año: "))
bisiesto = año % 4 == 0 and (año % 100 != 0 or año % 400 == 0)
print(f"{año} tiene {366 if bisiesto else 365} días")
```

## 11. Ecuación de segundo grado

Pedir los coeficientes de $ax^2+bx+c=0$ y mostrar sus soluciones reales, contemplando los casos de una solución, ninguna, infinitas o ecuación de primer grado.

```python
from math import sqrt

a = float(input("a: "))
b = float(input("b: "))
c = float(input("c: "))
if a == 0:
    # degenera en bx + c = 0
    if b != 0:
        print("Solución:", -c / b)
    else:
        print("Infinitas soluciones" if c == 0 else "Sin solución")
else:
    d = b**2 - 4 * a * c
    if d > 0:
        print("Soluciones:", (-b + sqrt(d)) / (2 * a), (-b - sqrt(d)) / (2 * a))
    elif d == 0:
        print("Solución doble:", -b / (2 * a))
    else:
        print("Sin solución real")
```
