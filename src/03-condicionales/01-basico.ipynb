# Condicionales · Básico

`if`, `elif`, `else` y operadores lógicos.

## 1. Mayoría de edad

Pedir la edad y mostrar si es mayor de edad.

```python
edad = int(input("Edad: "))
print("Mayor de edad" if edad >= 18 else "Menor de edad")
```

## 2. Contraseña sin distinguir mayúsculas

Guardar una contraseña en una variable, pedirla al usuario y decir si coincide sin tener en cuenta mayúsculas y minúsculas.

```python
contraseña = "secreto"
intento = input("Contraseña: ")
print("Correcta" if intento.lower() == contraseña.lower() else "Incorrecta")
```

## 3. División segura

Pedir dos números y mostrar su división, o un error si el divisor es cero.

```python
a = float(input("Dividendo: "))
b = float(input("Divisor: "))
print("Error: división por cero" if b == 0 else a / b)
```

## 4. Par o impar

Pedir un entero y decir si es par o impar.

```python
n = int(input("Entero: "))
print("Par" if n % 2 == 0 else "Impar")
```

## 5. ¿Tiene que tributar?

Tributa quien tiene más de 16 años e ingresos mensuales de al menos 1000 €. Pedir edad e ingresos y decir si tributa.

```python
edad = int(input("Edad: "))
ingresos = float(input("Ingresos mensuales: "))
print("Tiene que tributar" if edad > 16 and ingresos >= 1000 else "No tiene que tributar")
```

## 6. ¿Quién es más joven?

Leer la edad de dos personas y decir cuál es más joven, o si tienen la misma edad.

```python
a = int(input("Edad 1: "))
b = int(input("Edad 2: "))
if a < b:
    print("La primera es más joven")
elif b < a:
    print("La segunda es más joven")
else:
    print("Tienen la misma edad")
```

## 7. ¿Es paréntesis?

Leer un carácter y mostrar `Es paréntesis` solo si es `(` o `)`.

```python
c = input("Carácter: ")
if c in ("(", ")"):
    print("Es paréntesis")
```

## 8. Doble de un impar

Dado un entero, decir si es el doble de un número impar (14 es el doble de 7).

```python
n = int(input("Entero: "))
print(n % 2 == 0 and (n // 2) % 2 == 1)
```

## 9. Comparar con el cuadrado

Dados dos enteros, decir si el segundo es el cuadrado exacto del primero, menor o mayor que él.

```python
a = int(input("Primero: "))
b = int(input("Segundo: "))
if b == a**2:
    print("El segundo es el cuadrado exacto del primero")
elif b < a**2:
    print("El segundo es menor que el cuadrado del primero")
else:
    print("El segundo es mayor que el cuadrado del primero")
```

## 10. Mayúscula o minúscula

Leer un carácter y mostrar `Es una MAYÚSCULA` o `Es una MINÚSCULA` si es una letra del alfabeto inglés; en otro caso, nada.

```python
c = input("Carácter: ")
if "A" <= c <= "Z":
    print("Es una MAYÚSCULA")
elif "a" <= c <= "z":
    print("Es una MINÚSCULA")
```
