Leer y escribir ficheros de texto, leer de internet y gestionar ficheros con `os`.

## 1. Guardar una tabla de multiplicar

Escribir una función que pida un entero entre 1 y 10 y guarde su tabla de multiplicar en `tabla-n.txt`.

```python
def guarda_tabla():
    n = int(input("Número (1-10): "))
    with open(f"tabla-{n}.txt", "w", encoding="utf-8") as f:
        for i in range(1, 11):
            f.write(f"{n} x {i} = {n * i}\n")

guarda_tabla()
```

## 2. Leer una tabla de multiplicar

Escribir una función que pida un entero entre 1 y 10, lea `tabla-n.txt` y la muestre, o avise si el fichero no existe.

```python
def muestra_tabla():
    n = int(input("Número (1-10): "))
    try:
        with open(f"tabla-{n}.txt", encoding="utf-8") as f:
            print(f.read())
    except FileNotFoundError:
        print(f"No existe el fichero tabla-{n}.txt")

muestra_tabla()
```

## 3. Leer una línea concreta

Escribir una función que pida $n$ y $m$ entre 1 y 10, lea `tabla-n.txt` y muestre su línea $m$, o avise si el fichero no existe.

```python
def muestra_linea():
    n = int(input("Tabla (1-10): "))
    m = int(input("Línea (1-10): "))
    try:
        with open(f"tabla-{n}.txt", encoding="utf-8") as f:
            print(f.readlines()[m - 1], end="")
    except FileNotFoundError:
        print(f"No existe el fichero tabla-{n}.txt")

muestra_linea()
```

## 4. Contar palabras de un fichero de internet

Acceder a un fichero de internet por su URL y mostrar cuántas palabras contiene (por ejemplo, *El Quijote* en el Proyecto Gutenberg).

```python
from urllib import request
from urllib.error import URLError

def contar_palabras(url):
    try:
        with request.urlopen(url) as f:
            return len(f.read().decode("utf-8").split())
    except (URLError, ValueError) as e:
        return f"Error al acceder a {url}: {e}"

print(contar_palabras("https://www.gutenberg.org/cache/epub/2000/pg2000.txt"))
print(contar_palabras("https://no-existe.txt"))
```

## 5. Crear, renombrar y borrar ficheros con `os`

Crear varios ficheros vacíos en una carpeta y renombrarlos añadiendo un prefijo con su orden alfabético (`archivo.txt` → `1_archivo.txt`, `ejemplo.txt` → `2_ejemplo.txt`). Al final, borrar ficheros y carpeta.

```python
import os

ruta = "ejemplo-ficheros"
os.makedirs(ruta, exist_ok=True)
for nombre in ("ejemplo.txt", "archivo.txt", "notas.txt", "datos.txt"):
    open(os.path.join(ruta, nombre), "w").close()

def renombra_por_orden(ruta):
    for i, nombre in enumerate(sorted(os.listdir(ruta)), 1):
        os.rename(os.path.join(ruta, nombre), os.path.join(ruta, f"{i}_{nombre}"))

renombra_por_orden(ruta)
print(sorted(os.listdir(ruta)))

for nombre in os.listdir(ruta):
    os.remove(os.path.join(ruta, nombre))
os.rmdir(ruta)
print(os.path.exists(ruta))
```
