## 1. Clase `Statistics`

Crear una clase `Statistics` que reciba una muestra y tenga métodos para `count`, `sum`, `min`, `max`, `range`, `mean`, `median`, `mode`, `var`, `std`, `freq_dist` (porcentaje de cada valor, de más a menos frecuente) y `describe`, que muestre todo.

```python
from collections import Counter

class Statistics:
    def __init__(self, datos):
        self.datos = list(datos)

    def count(self):
        return len(self.datos)

    def sum(self):
        return sum(self.datos)

    def min(self):
        return min(self.datos)

    def max(self):
        return max(self.datos)

    def range(self):
        return self.max() - self.min()

    def mean(self):
        return self.sum() / self.count()

    def median(self):
        d, n = sorted(self.datos), self.count()
        return d[n // 2] if n % 2 else (d[n // 2 - 1] + d[n // 2]) / 2

    def mode(self):
        valor, veces = Counter(self.datos).most_common(1)[0]
        return {"mode": valor, "count": veces}

    def var(self):
        m = self.mean()
        return sum((x - m) ** 2 for x in self.datos) / self.count()

    def std(self):
        return self.var() ** 0.5

    def freq_dist(self):
        return [(v * 100 / self.count(), k) for k, v in Counter(self.datos).most_common()]

    def describe(self):
        return "\n".join([
            f"Count: {self.count()}", f"Sum: {self.sum()}", f"Min: {self.min()}", f"Max: {self.max()}",
            f"Range: {self.range()}", f"Mean: {self.mean():.0f}", f"Median: {self.median()}",
            f"Mode: {tuple(self.mode().values())}", f"Variance: {self.var():.1f}",
            f"Standard Deviation: {self.std():.1f}", f"Frequency Distribution: {self.freq_dist()}",
        ])

ages = [31, 26, 34, 37, 27, 26, 32, 32, 26, 27, 27, 24, 32, 33, 27, 25, 26, 38, 37, 31, 34, 24, 33, 29, 26]
data = Statistics(ages)
print(data.describe())
```

## 2. Clase `PersonAccount`

Crear una clase `PersonAccount` con nombre, apellido, ingresos y gastos (cada uno con su descripción) y los métodos `add_income`, `add_expense`, `total_income`, `total_expense`, `account_balance` y `account_info`.

```python
class PersonAccount:
    def __init__(self, firstname, lastname):
        self.firstname = firstname
        self.lastname = lastname
        self.incomes = []
        self.expenses = []

    def add_income(self, cantidad, descripcion):
        self.incomes.append((cantidad, descripcion))

    def add_expense(self, cantidad, descripcion):
        self.expenses.append((cantidad, descripcion))

    def total_income(self):
        return sum(c for c, _ in self.incomes)

    def total_expense(self):
        return sum(c for c, _ in self.expenses)

    def account_balance(self):
        return self.total_income() - self.total_expense()

    def account_info(self):
        return (f"{self.firstname} {self.lastname}: ingresos {self.total_income()}, "
                f"gastos {self.total_expense()}, saldo {self.account_balance()}")

cuenta = PersonAccount("Ana", "Pérez")
cuenta.add_income(2000, "nómina")
cuenta.add_income(150, "clases particulares")
cuenta.add_expense(800, "alquiler")
cuenta.add_expense(250, "comida")
print(cuenta.account_info())
```

## 4. Herencia y `super()`

Crear una clase `Empleado` (nombre, sueldo, método `salario` y `__str__`) y una subclase `Gerente` que añada un plus y un equipo, reutilizando el constructor y el `salario` del padre con `super()`. Comprobar `isinstance` e `issubclass`. Después, con herencia múltiple en rombo, mostrar que `super()` sigue el orden de resolución de métodos (`__mro__`) y no simplemente la clase padre.

```python
class Empleado:
    def __init__(self, nombre, sueldo):
        self.nombre = nombre
        self.sueldo = sueldo

    def salario(self):
        return self.sueldo

    def __str__(self):
        return f"{type(self).__name__} {self.nombre}: {self.salario()} €"

class Gerente(Empleado):
    def __init__(self, nombre, sueldo, plus):
        super().__init__(nombre, sueldo)
        self.plus = plus
        self.equipo = []

    def contrata(self, empleado):
        self.equipo.append(empleado)

    def salario(self):
        return super().salario() + self.plus + 100 * len(self.equipo)

ana = Empleado("Ana", 1800)
luis = Gerente("Luis", 2500, 600)
luis.contrata(ana)
print(ana)
print(luis)
print(isinstance(luis, Empleado), isinstance(ana, Gerente), issubclass(Gerente, Empleado), issubclass(Gerente, object))
```

## 7. `@property` con validación

Crear una clase `Temperatura` cuyo atributo `celsius` sea una propiedad que rechace valores que no sean números o que estén por debajo del cero absoluto (−273.15 °C), una propiedad `fahrenheit` calculada que también se pueda asignar y una propiedad `kelvin` de solo lectura. El constructor debe pasar por la validación.

```python
class Temperatura:
    CERO_ABSOLUTO = -273.15

    def __init__(self, celsius=0.0):
        self.celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, valor):
        if not isinstance(valor, (int, float)):
            raise TypeError(f"la temperatura debe ser un número, no {valor!r}")
        if valor < self.CERO_ABSOLUTO:
            raise ValueError(f"{valor} °C está por debajo del cero absoluto")
        self._celsius = valor

    @property
    def fahrenheit(self):
        return self.celsius * 9 / 5 + 32

    @fahrenheit.setter
    def fahrenheit(self, valor):
        self.celsius = (valor - 32) * 5 / 9

    @property
    def kelvin(self):
        return self.celsius - self.CERO_ABSOLUTO

t = Temperatura(25)
print(t.celsius, t.fahrenheit, round(t.kelvin, 2))
t.fahrenheit = 212
print(t.celsius)
for mala in (-300, "caliente"):
    try:
        Temperatura(mala)
    except (ValueError, TypeError) as e:
        print(f"{type(e).__name__}: {e}")
try:
    t.kelvin = 0
except AttributeError as e:
    print("AttributeError:", e)
```

## 8. `classmethod` y `staticmethod`

Crear una clase `Fecha` con dos constructores alternativos como `classmethod`: `desde_cadena('dd/mm/aaaa')` y `desde_iso('aaaa-mm-dd')`, y los `staticmethod` `es_bisiesto(año)` y `dias_mes(mes, año)`, usados para validar. Comprobar que el constructor alternativo de una subclase crea objetos de la subclase y comparar `es_bisiesto` con `calendar.isleap`.

```python
import calendar

class Fecha:
    DIAS = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

    def __init__(self, dia, mes, año):
        if not 1 <= mes <= 12 or not 1 <= dia <= self.dias_mes(mes, año):
            raise ValueError(f"fecha no válida: {dia}/{mes}/{año}")
        self.dia, self.mes, self.año = dia, mes, año

    @classmethod
    def desde_cadena(cls, texto):
        dia, mes, año = map(int, texto.split("/"))
        return cls(dia, mes, año)

    @classmethod
    def desde_iso(cls, texto):
        año, mes, dia = map(int, texto.split("-"))
        return cls(dia, mes, año)

    @staticmethod
    def es_bisiesto(año):
        return año % 4 == 0 and (año % 100 != 0 or año % 400 == 0)

    @staticmethod
    def dias_mes(mes, año):
        return 29 if mes == 2 and Fecha.es_bisiesto(año) else Fecha.DIAS[mes - 1]

    def __str__(self):
        return f"{self.dia:02}/{self.mes:02}/{self.año}"

class FechaLarga(Fecha):
    MESES = ["enero", "febrero", "marzo", "abril", "mayo", "junio", "julio",
             "agosto", "septiembre", "octubre", "noviembre", "diciembre"]

    def __str__(self):
        return f"{self.dia} de {self.MESES[self.mes - 1]} de {self.año}"

print(Fecha.desde_cadena("6/11/2026"), Fecha.desde_iso("2024-02-29"))
f = FechaLarga.desde_iso("2026-05-30")
print(type(f).__name__, f)
print([a for a in (1900, 2000, 2024, 2026) if Fecha.es_bisiesto(a)])
assert all(Fecha.es_bisiesto(a) == calendar.isleap(a) for a in range(1, 3001))
try:
    Fecha.desde_cadena("29/02/2026")
except ValueError as e:
    print("ValueError:", e)
```

## 9. `__repr__`, `__str__`, `__eq__` y `__hash__`

Crear una clase `Punto` con `__repr__` (pensado para programadores: permite recrear el objeto con `eval`), `__str__` (pensado para usuarios), y `__eq__` y `__hash__` coherentes, de modo que dos puntos iguales cuenten como uno en un conjunto y sirvan como la misma clave de un diccionario. Mostrar qué pasa si no se define ninguno y si solo se define `__eq__`.

```python
class Punto:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __repr__(self):
        return f"Punto({self.x!r}, {self.y!r})"

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __eq__(self, otro):
        if not isinstance(otro, Punto):
            return NotImplemented
        return (self.x, self.y) == (otro.x, otro.y)

    def __hash__(self):
        return hash((self.x, self.y))

p, q = Punto(1, 2), Punto(1, 2)
print(p, repr(p), [p, q])
print(p == q, p is q, p == (1, 2), eval(repr(p)) == p)
print(len({p, q, Punto(3, 4)}), {p: "tesoro"}[q])
```

## 10. Operadores y orden: `__add__`, `__mul__`, `__lt__` y `total_ordering`

Crear una clase `Fraccion` siempre simplificada (con `math.gcd` y el signo en el numerador) que admita `+` y `*`, también con enteros a cualquier lado (`__radd__`, `__rmul__`), y todas las comparaciones definiendo solo `__eq__` y `__lt__` con `functools.total_ordering`. Comprobar que `sum` y `sorted` funcionan y comparar los resultados con `fractions.Fraction` en muchos casos aleatorios.

```python
from functools import total_ordering
from math import gcd

@total_ordering
class Fraccion:
    def __init__(self, num, den=1):
        if den == 0:
            raise ZeroDivisionError("denominador cero")
        if den < 0:
            num, den = -num, -den
        m = gcd(num, den)
        self.num, self.den = num // m, den // m

    @staticmethod
    def _convierte(x):
        return Fraccion(x) if isinstance(x, int) else x

    def __add__(self, otra):
        otra = self._convierte(otra)
        if not isinstance(otra, Fraccion):
            return NotImplemented
        return Fraccion(self.num * otra.den + otra.num * self.den, self.den * otra.den)

    __radd__ = __add__

    def __mul__(self, otra):
        otra = self._convierte(otra)
        if not isinstance(otra, Fraccion):
            return NotImplemented
        return Fraccion(self.num * otra.num, self.den * otra.den)

    __rmul__ = __mul__

    def __eq__(self, otra):
        otra = self._convierte(otra)
        if not isinstance(otra, Fraccion):
            return NotImplemented
        return (self.num, self.den) == (otra.num, otra.den)

    def __lt__(self, otra):
        otra = self._convierte(otra)
        if not isinstance(otra, Fraccion):
            return NotImplemented
        return self.num * otra.den < otra.num * self.den

    def __hash__(self):
        return hash((self.num, self.den))

    def __repr__(self):
        return f"{self.num}/{self.den}" if self.den != 1 else str(self.num)

a, b = Fraccion(1, 2), Fraccion(2, 6)
print(a + b, a * b, 2 * a, 1 + b, Fraccion(3, -6))
print(a > b, a >= Fraccion(2, 4), a != b, Fraccion(4, 2) == 2)
print(sorted([a, b, Fraccion(-3, 4), Fraccion(5)]), sum(Fraccion(1, n) for n in range(1, 5)))
```

```python
import random
from fractions import Fraction

random.seed(1)
for _ in range(1000):
    n1, n2 = random.randint(-20, 20), random.randint(-20, 20)
    d1, d2 = random.randint(1, 9) * random.choice((1, -1)), random.randint(1, 9) * random.choice((1, -1))
    x, y, fx, fy = Fraccion(n1, d1), Fraccion(n2, d2), Fraction(n1, d1), Fraction(n2, d2)
    for mia, ref in ((x + y, fx + fy), (x * y, fx * fy), (3 + x, 3 + fx)):
        assert (mia.num, mia.den) == (ref.numerator, ref.denominator)
    assert (x < y, x <= y, x > y, x >= y, x == y) == (fx < fy, fx <= fy, fx > fy, fx >= fy, fx == fy)
print("Coincide con fractions.Fraction en 1000 casos")
```

## 11. Protocolo de contenedor: `__len__`, `__getitem__`, `__iter__` y `__contains__`

Crear una clase `Baraja` de cartas españolas (40 cartas, cada una una tupla `(valor, palo)`) con `__len__`, `__getitem__` (que admita índices y cortes), `__iter__` y `__contains__`, y un método para barajar con semilla. Comprobar que funcionan `len`, `in`, `for`, los cortes, `reversed` y `random.choice`. Mostrar también que con solo `__getitem__` Python ya sabe recorrer un objeto y usar `in`.

```python
import random

class Baraja:
    PALOS = ("oros", "copas", "espadas", "bastos")
    VALORES = (1, 2, 3, 4, 5, 6, 7, 10, 11, 12)

    def __init__(self):
        self._cartas = [(v, p) for p in self.PALOS for v in self.VALORES]

    def __len__(self):
        return len(self._cartas)

    def __getitem__(self, i):
        return self._cartas[i]

    def __iter__(self):
        return iter(self._cartas)

    def __contains__(self, carta):
        return carta in self._cartas

    def baraja(self, semilla=None):
        random.Random(semilla).shuffle(self._cartas)

b = Baraja()
print(len(b), b[0], b[-1], b[:3])
print((7, "oros") in b, (8, "copas") in b)
print(list(reversed(b))[:2])
random.seed(3)
print(random.choice(b))
b.baraja(semilla=3)
print(b[:5], sum(1 for _, palo in b if palo == "espadas"))
```

## 12. Clases abstractas con `abc`

Crear una clase abstracta `Figura` (con `abc.ABC`) con los métodos abstractos `area` y `perimetro` y un método concreto `describe`. Comprobar que no se puede instanciar `Figura` ni una subclase que no implemente todos los métodos abstractos. Crear `Circulo`, `Rectangulo` y `Triangulo` (área con la fórmula de Herón, validando la desigualdad triangular).

```python
import math
from abc import ABC, abstractmethod

class Figura(ABC):
    @abstractmethod
    def area(self): ...

    @abstractmethod
    def perimetro(self): ...

    def describe(self):
        return f"{type(self).__name__}: área {self.area():.2f}, perímetro {self.perimetro():.2f}"

class Circulo(Figura):
    def __init__(self, r):
        self.r = r

    def area(self):
        return math.pi * self.r ** 2

    def perimetro(self):
        return 2 * math.pi * self.r

class Rectangulo(Figura):
    def __init__(self, base, altura):
        self.base, self.altura = base, altura

    def area(self):
        return self.base * self.altura

    def perimetro(self):
        return 2 * (self.base + self.altura)

class Triangulo(Figura):
    def __init__(self, a, b, c):
        if 2 * max(a, b, c) >= a + b + c:
            raise ValueError(f"({a}, {b}, {c}) no cumple la desigualdad triangular")
        self.lados = (a, b, c)

    def perimetro(self):
        return sum(self.lados)

    def area(self):
        s = self.perimetro() / 2
        a, b, c = self.lados
        return math.sqrt(s * (s - a) * (s - b) * (s - c))

class Incompleta(Figura):
    def area(self):
        return 0

for clase in (Figura, Incompleta):
    try:
        clase()
    except TypeError as e:
        print("TypeError:", e)
try:
    Triangulo(1, 2, 3)
except ValueError as e:
    print("ValueError:", e)

figuras = [Circulo(1), Rectangulo(3, 4), Triangulo(3, 4, 5)]
for f in figuras:
    print(f.describe())
assert math.isclose(Triangulo(3, 4, 5).area(), 6)
print(all(isinstance(f, Figura) for f in figuras), sorted(Figura.__abstractmethods__))
```

- `Banco`, que *contiene* las cuentas (composición) e implementa `len`, `in`, `for` y acceso por número, transferencias, el cierre mensual polimórfico y el saldo total.

Si una retirada supera lo disponible, lanzar una excepción propia `SaldoInsuficiente`.

```python
from abc import ABC, abstractmethod

class SaldoInsuficiente(Exception):
    pass

class Cuenta(ABC):
    _siguiente = 1

    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.numero = f"ES{Cuenta._siguiente:04}"
        Cuenta._siguiente += 1
        self._saldo = 0
        self.movimientos = []
        if saldo:
            self.ingresa(saldo)

    @classmethod
    def desde_texto(cls, texto):
        titular, saldo = texto.split(";")
        return cls(titular.strip(), float(saldo))

    @staticmethod
    def _valida(cantidad):
        if not isinstance(cantidad, (int, float)) or cantidad <= 0:
            raise ValueError(f"cantidad no válida: {cantidad!r}")
        return cantidad

    @property
    def saldo(self):
        return self._saldo

    def disponible(self):
        return self._saldo

    def ingresa(self, cantidad):
        self._saldo = round(self._saldo + self._valida(cantidad), 2)
        self.movimientos.append(("ingreso", cantidad))

    def retira(self, cantidad):
        if self._valida(cantidad) > self.disponible():
            raise SaldoInsuficiente(f"{self.numero}: no se pueden retirar {cantidad} €")
        self._saldo = round(self._saldo - cantidad, 2)
        self.movimientos.append(("retirada", cantidad))

    @abstractmethod
    def fin_de_mes(self): ...

    def __repr__(self):
        return f"{type(self).__name__}({self.titular!r}, {self.saldo})"

    def __str__(self):
        return f"{self.numero} {type(self).__name__:16} {self.titular:5} {self.saldo:>9.2f} €"

    def __eq__(self, otra):
        return isinstance(otra, Cuenta) and self.numero == otra.numero

    def __hash__(self):
        return hash(self.numero)

class CuentaCorriente(Cuenta):
    COMISION = 5

    def __init__(self, titular, saldo=0, descubierto=200):
        super().__init__(titular, saldo)
        self.descubierto = descubierto

    def disponible(self):
        return super().disponible() + self.descubierto

    def fin_de_mes(self):
        if self.saldo < 1000:
            self._saldo = round(self._saldo - self.COMISION, 2)
            self.movimientos.append(("comisión", self.COMISION))

class CuentaAhorro(Cuenta):
    def __init__(self, titular, saldo=0, interes=0.02):
        super().__init__(titular, saldo)
        self.interes = interes

    def fin_de_mes(self):
        intereses = round(self.saldo * self.interes / 12, 2)
        if intereses > 0:
            self.ingresa(intereses)

class Banco:
    def __init__(self, nombre):
        self.nombre = nombre
        self._cuentas = {}

    def abre(self, cuenta):
        self._cuentas[cuenta.numero] = cuenta
        return cuenta

    def __len__(self):
        return len(self._cuentas)

    def __iter__(self):
        return iter(self._cuentas.values())

    def __getitem__(self, numero):
        return self._cuentas[numero]

    def __contains__(self, cuenta):
        return getattr(cuenta, "numero", cuenta) in self._cuentas

    def transfiere(self, origen, destino, cantidad):
        self[origen].retira(cantidad)
        self[destino].ingresa(cantidad)

    def cierre_mensual(self):
        for cuenta in self:
            cuenta.fin_de_mes()

    def total(self):
        return round(sum(c.saldo for c in self), 2)
```

```python
banco = Banco("Banco Insular")
ana = banco.abre(CuentaCorriente("Ana", 1200))
luis = banco.abre(CuentaAhorro.desde_texto("Luis; 3000"))
eva = banco.abre(CuentaCorriente.desde_texto("Eva;150"))

banco.transfiere(eva.numero, ana.numero, 300)
try:
    banco.transfiere(eva.numero, luis.numero, 100)
except SaldoInsuficiente as e:
    print("SaldoInsuficiente:", e)
for mala in (-5, "diez"):
    try:
        ana.ingresa(mala)
    except ValueError as e:
        print("ValueError:", e)
try:
    ana.saldo = 1_000_000
except AttributeError:
    print("El saldo es de solo lectura")

total = banco.total()
banco.cierre_mensual()
assert banco.total() == total - CuentaCorriente.COMISION + 5.0
for cuenta in sorted(banco, key=lambda c: c.saldo, reverse=True):
    print(cuenta)
print(len(banco), ana in banco, "ES9999" in banco, banco[luis.numero] is luis, repr(eva))
print(eva.movimientos)
```
