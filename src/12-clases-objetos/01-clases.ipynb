## 1. Clase `Statistics`

Crear una clase `Statistics` que reciba una muestra y tenga métodos para `count`, `sum`, `min`, `max`, `range`, `mean`, `median`, `mode`, `var`, `std`, `freq_dist` (porcentaje de cada valor, de más a menos frecuente) y `describe`, que muestre todo.

```python
from collections import Counter

class Statistics:
    def __init__(self, datos):
        self.datos = list(datos)

    def count(self):
        return len(self.datos)

    def sum(self):
        return sum(self.datos)

    def min(self):
        return min(self.datos)

    def max(self):
        return max(self.datos)

    def range(self):
        return self.max() - self.min()

    def mean(self):
        return self.sum() / self.count()

    def median(self):
        d, n = sorted(self.datos), self.count()
        return d[n // 2] if n % 2 else (d[n // 2 - 1] + d[n // 2]) / 2

    def mode(self):
        valor, veces = Counter(self.datos).most_common(1)[0]
        return {"mode": valor, "count": veces}

    def var(self):
        m = self.mean()
        return sum((x - m) ** 2 for x in self.datos) / self.count()

    def std(self):
        return self.var() ** 0.5

    def freq_dist(self):
        return [(v * 100 / self.count(), k) for k, v in Counter(self.datos).most_common()]

    def describe(self):
        return "\n".join([
            f"Count: {self.count()}", f"Sum: {self.sum()}", f"Min: {self.min()}", f"Max: {self.max()}",
            f"Range: {self.range()}", f"Mean: {self.mean():.0f}", f"Median: {self.median()}",
            f"Mode: {tuple(self.mode().values())}", f"Variance: {self.var():.1f}",
            f"Standard Deviation: {self.std():.1f}", f"Frequency Distribution: {self.freq_dist()}",
        ])

ages = [31, 26, 34, 37, 27, 26, 32, 32, 26, 27, 27, 24, 32, 33, 27, 25, 26, 38, 37, 31, 34, 24, 33, 29, 26]
data = Statistics(ages)
print(data.describe())
```

## 2. Clase `PersonAccount`

Crear una clase `PersonAccount` con nombre, apellido, ingresos y gastos (cada uno con su descripción) y los métodos `add_income`, `add_expense`, `total_income`, `total_expense`, `account_balance` y `account_info`.

```python
class PersonAccount:
    def __init__(self, firstname, lastname):
        self.firstname = firstname
        self.lastname = lastname
        self.incomes = []
        self.expenses = []

    def add_income(self, cantidad, descripcion):
        self.incomes.append((cantidad, descripcion))

    def add_expense(self, cantidad, descripcion):
        self.expenses.append((cantidad, descripcion))

    def total_income(self):
        return sum(c for c, _ in self.incomes)

    def total_expense(self):
        return sum(c for c, _ in self.expenses)

    def account_balance(self):
        return self.total_income() - self.total_expense()

    def account_info(self):
        return (f"{self.firstname} {self.lastname}: ingresos {self.total_income()}, "
                f"gastos {self.total_expense()}, saldo {self.account_balance()}")

cuenta = PersonAccount("Ana", "Pérez")
cuenta.add_income(2000, "nómina")
cuenta.add_income(150, "clases particulares")
cuenta.add_expense(800, "alquiler")
cuenta.add_expense(250, "comida")
print(cuenta.account_info())
```
