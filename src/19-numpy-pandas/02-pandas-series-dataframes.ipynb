# Pandas · Series y DataFrames

Crear y operar con `Series` y `DataFrame` de pandas.

## 1. Ventas con descuento

Pedir las ventas de un rango de años y mostrar una serie indexada por año antes y después de aplicar un 10 % de descuento.

```python
import pandas as pd

inicio, fin = int(input("Año inicial: ")), int(input("Año final: "))
ventas = pd.Series({año: float(input(f"Ventas de {año}: ")) for año in range(inicio, fin + 1)})
print(ventas)
print(ventas * 0.9)
```

## 2. Estadísticos de notas

Función que reciba un diccionario de notas y devuelva una serie con la nota mínima, máxima, media y la desviación típica.

```python
notas = {"Juan": 9, "María": 6.5, "Pedro": 4, "Carmen": 8.5, "Luis": 5}

def estadisticos(notas):
    s = pd.Series(notas)
    return pd.Series({"Mínimo": s.min(), "Máximo": s.max(), "Media": s.mean(), "Desviación típica": s.std()})

print(estadisticos(notas))
```

## 3. Aprobados ordenados

Función que reciba un diccionario de notas y devuelva una serie con los aprobados ordenados de mayor a menor nota.

```python
def aprobados(notas):
    s = pd.Series(notas)
    return s[s >= 5].sort_values(ascending=False)

print(aprobados(notas))
```

## 4. DataFrame de ventas y gastos

Crear y mostrar un DataFrame con las ventas y gastos de enero a abril.

```python
datos = pd.DataFrame({"Mes": ["Enero", "Febrero", "Marzo", "Abril"],
                      "Ventas": [30500, 35600, 28300, 33900],
                      "Gastos": [22000, 23400, 18100, 20700]})
print(datos)
```

## 5. Balance de unos meses

Función que reciba ese DataFrame y una lista de meses y devuelva el balance (ventas − gastos) total en esos meses.

```python
def balance(datos, meses):
    seleccion = datos[datos["Mes"].isin(meses)]
    return (seleccion["Ventas"] - seleccion["Gastos"]).sum()

print(balance(datos, ["Enero", "Marzo"]))
```
