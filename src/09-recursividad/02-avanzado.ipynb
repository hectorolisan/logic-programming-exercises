## 1. Fibonacci

Calcular $F_{12}$ con $F_1 = F_2 = 1$ y $F_n = F_{n-1} + F_{n-2}$. La versión ingenua repite muchas llamadas; comparar con una versión memoizada.

```python
from functools import cache

def fibonacci(n):
    return 1 if n <= 2 else fibonacci(n - 1) + fibonacci(n - 2)

@cache
def fibonacci_memo(n):
    return 1 if n <= 2 else fibonacci_memo(n - 1) + fibonacci_memo(n - 2)

print(fibonacci(12), fibonacci_memo(12), fibonacci_memo(100))
```

## 2. Números combinatorios

Con $\binom{n}{m} = \binom{n-1}{m} + \binom{n-1}{m-1}$ y $\binom{n}{0} = \binom{n}{n} = 1$, leer $n$ y mostrar $\binom{n}{m}$ para $m$ entre 0 y $n$ usando una función recursiva `combinaciones`.

```python
def combinaciones(n, m):
    if m == 0 or m == n:
        return 1
    return combinaciones(n - 1, m) + combinaciones(n - 1, m - 1)

n = int(input("n: "))
print([combinaciones(n, m) for m in range(n + 1)])
```

## 3. Particiones de un conjunto

El número de formas de dividir un conjunto de $n$ elementos en $k$ subconjuntos no vacíos (Stirling de segunda especie) cumple $S(n,k) = S(n-1,k-1) + k\,S(n-1,k)$, con $S(n,1) = S(n,n) = 1$. Leer $n$ y mostrar $S(n,m)$ para $m$ entre 0 y $n$ con una función recursiva `particiones`.

```python
def particiones(n, k):
    if k == 0:
        return 1 if n == 0 else 0
    if k == 1 or k == n:
        return 1
    return particiones(n - 1, k - 1) + k * particiones(n - 1, k)

n = int(input("n: "))
print([particiones(n, m) for m in range(n + 1)])
```

## 4. MCD recursivo con traza

Definir el MCD de Euclides de forma recursiva ($\text{mcd}(a, 0) = a$, $\text{mcd}(a, b) = \text{mcd}(b, a \bmod b)$) mostrando cada llamada, y trazarlo para 1470 y 693, y para 323 y 323.

```python
def mcd(a, b, nivel=0):
    print("  " * nivel + f"mcd({a}, {b})")
    return a if b == 0 else mcd(b, a % b, nivel + 1)

print(mcd(1470, 693))
print(mcd(323, 323))
```

## 5. Torres de Hanói

Mover $n$ discos de la torre 1 a la 3 usando la 2 como auxiliar, sin poner nunca un disco sobre otro más pequeño. Mostrar los movimientos para 3 discos, escribir `movimientos_hanoi(n)` y explicar por qué no conviene ejecutar `resuelve_hanoi(64, 1, 3, 2)`.

```python
def resuelve_hanoi(n, origen, destino, auxiliar):
    if n == 0:
        return
    resuelve_hanoi(n - 1, origen, auxiliar, destino)
    print(f"Mueve disco {n} de {origen} a {destino}")
    resuelve_hanoi(n - 1, auxiliar, destino, origen)

def movimientos_hanoi(n):
    return 0 if n == 0 else 2 * movimientos_hanoi(n - 1) + 1

resuelve_hanoi(3, 1, 3, 2)
print([movimientos_hanoi(n) for n in range(1, 11)])
# 2^64 - 1 movimientos: a uno por segundo, unos 585 000 millones de años
print(f"64 discos: {movimientos_hanoi(64):,} movimientos ≈ {movimientos_hanoi(64) / (3600 * 24 * 365.25):.3e} años")
```
