# Recursividad · Básico

Funciones que se llaman a sí mismas: caso base y caso recursivo.

## 1. Factorial recursivo

Definir `factorial` de forma recursiva ($0! = 1$, $n! = n \cdot (n-1)!$) y mostrar su valor para $n$ entre 0 y 7.

```python
def factorial(n):
    return 1 if n == 0 else n * factorial(n - 1)

for n in range(8):
    print(f"{n}! = {factorial(n)}")
```

## 2. Suma de los n primeros naturales

Calcular recursivamente $\sum_{i=1}^{n} i$, sabiendo que vale 1 si $n = 1$ y $n + \sum_{i=1}^{n-1} i$ si $n > 1$.

```python
def suma(n):
    return 1 if n == 1 else n + suma(n - 1)

print(suma(100))
```

## 3. Sumatorio entre m y n

Calcular recursivamente $\sum_{i=m}^{n} i$.

```python
def sumatorio(m, n):
    return 0 if m > n else m + sumatorio(m + 1, n)

print(sumatorio(3, 10))
```

## 4. Número de dígitos

Calcular recursivamente cuántos dígitos tiene un entero en base 10.

```python
def digitos(n):
    n = abs(n)
    return 1 if n < 10 else 1 + digitos(n // 10)

print(digitos(0), digitos(7), digitos(-12345))
```

## 5. Potencia con exponente negativo

Calcular recursivamente $b^e$ para cualquier exponente entero, usando que $b^{-e} = 1 / b^{e}$.

```python
def potencia(b, e):
    if e < 0:
        return 1 / potencia(b, -e)
    return 1 if e == 0 else b * potencia(b, e - 1)

print(potencia(2, 10), potencia(2, -3))
```
