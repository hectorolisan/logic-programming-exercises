## 1. Factorizar en primos

Descomponer un número en sus factores primos. Ejemplo: 765 = 3·3·5·17.

```python
n = int(input("Número: "))
factores, d = [], 2
while d * d <= n:
    while n % d == 0:
        factores.append(d)
        n //= d
    d += 1
if n > 1:
    factores.append(n)
print(" * ".join(map(str, factores)))
```

## 2. Primo de 8 dígitos

Escribir una función que encuentre un número primo muy grande, de 8 dígitos (el mayor).

```python
def es_primo(n):
    return n > 1 and all(n % d for d in range(2, int(n**0.5) + 1))

def mayor_primo(digitos):
    n = 10**digitos - 1
    while not es_primo(n):
        n -= 1
    return n

print(mayor_primo(8))
```

## 3. Fuerza bruta con `itertools`

Calcular cuántas contraseñas de longitud 10 se pueden formar con minúsculas y dígitos, y romper por fuerza bruta una contraseña probando todas las combinaciones (con una longitud corta para que acabe en un tiempo razonable).

```python
import itertools
import string

caracteres = string.ascii_lowercase + string.digits
print(f"Combinaciones de longitud 10: {len(caracteres) ** 10:,}")

def fuerza_bruta(objetivo, max_longitud):
    intentos = 0
    for longitud in range(1, max_longitud + 1):
        for combinacion in itertools.product(caracteres, repeat=longitud):
            intentos += 1
            if "".join(combinacion) == objetivo:
                return intentos
    return None

print("Rota en", fuerza_bruta("z9a", 3), "intentos")
```

## 4. Barra de progreso

Escribir una cuenta atrás con barra de progreso que se actualice más de una vez por segundo.

```python
import time

def cuenta_atras(segundos, pasos_por_segundo=4):
    total = segundos * pasos_por_segundo
    for paso in range(total + 1):
        progreso = paso * 20 // total
        restante = (total - paso) / pasos_por_segundo
        # \r reescribe la misma línea
        print(f"\r[{'#' * progreso}{'.' * (20 - progreso)}] {restante:4.2f}s", end="")
        time.sleep(1 / pasos_por_segundo)
    print()

cuenta_atras(2)
```
