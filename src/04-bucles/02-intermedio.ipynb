Acumuladores, bucles anidados, figuras y números primos.

## 1. Capital año a año

Pedir cantidad a invertir, interés anual (%) y años, y mostrar el capital obtenido al final de cada año.

````python
capital = float(input("Cantidad: "))
interes = float(input("Interés (%): "))
años = int(input("Años: "))
for año in range(1, años + 1):
    capital *= 1 + interes / 100
    print(f"Año {año}: {capital:.2f}")
````

## 2. Triángulo de asteriscos

Pedir una altura y dibujar un triángulo rectángulo de asteriscos.

````python
n = int(input("Altura: "))
for i in range(1, n + 1):
    print("*" * i)
````

## 3. Tablas de multiplicar

Mostrar las tablas de multiplicar del 1 al 10.

````python
for i in range(1, 11):
    print(" ".join(f"{i * j:3}" for j in range(1, 11)))
````

## 4. Triángulo de impares

Pedir una altura y mostrar un triángulo en el que cada fila lista los impares en orden descendente:

```
1
3 1
5 3 1
```

````python
n = int(input("Altura: "))
for i in range(1, 2 * n, 2):
    print(*range(i, 0, -2))
````

## 5. ¿Es primo?

Pedir un entero y decir si es primo.

````python
n = int(input("Entero: "))
# basta probar divisores hasta la raíz
primo = n > 1 and all(n % d for d in range(2, int(n**0.5) + 1))
print("Es primo" if primo else "No es primo")
````

## 6. Letras al revés

Pedir una palabra y mostrar sus letras una a una empezando por la última.

````python
for letra in reversed(input("Palabra: ")):
    print(letra)
````

## 7. Contar una letra

Pedir una frase y una letra y mostrar cuántas veces aparece la letra, recorriendo la frase con un bucle.

````python
frase = input("Frase: ")
letra = input("Letra: ")
veces = 0
for c in frase:
    if c == letra:
        veces += 1
print(veces)
````

## 8. Raíces n-ésimas

Leer un número y mostrar su raíz n-ésima para $n$ entre 2 y 100.

````python
x = float(input("Número: "))
for n in range(2, 101):
    print(f"raíz {n}: {x ** (1 / n):.6f}")
````

## 9. Máximo hasta un negativo

Leer números hasta que se introduzca uno negativo y mostrar el mayor de los leídos.

````python
maximo = None
while (n := float(input("Número: "))) >= 0:
    if maximo is None or n > maximo:
        maximo = n
print("Mayor:", maximo)
````

## 10. MCD de dos números

Calcular el máximo común divisor de dos enteros positivos probando divisores (sin `math.gcd`).

````python
a = int(input("a: "))
b = int(input("b: "))
for d in range(min(a, b), 0, -1):
    if a % d == 0 and b % d == 0:
        print("MCD:", d)
        break
````

## 11. MCD de tres números

Calcular el máximo común divisor de tres enteros positivos.

````python
a = int(input("a: "))
b = int(input("b: "))
c = int(input("c: "))
d = min(a, b, c)
while a % d or b % d or c % d:
    d -= 1
print("MCD:", d)
````
