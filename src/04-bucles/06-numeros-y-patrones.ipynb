## 1. FizzBuzz y par o impar

Recorrer del 1 al 100 escribiendo `Fizz` en los múltiplos de 3, `Buzz` en los de 5, `FizzBuzz` en los de ambos y el número en el resto. Obtener la misma lista con una sola comprensión y contar cuántos números se escriben tal cual. Después, decir si son pares o impares las alturas (en metros) de cinco picos canarios.

```python
fizzbuzz = []
for i in range(1, 101):
    if i % 15 == 0:
        fizzbuzz.append("FizzBuzz")
    elif i % 3 == 0:
        fizzbuzz.append("Fizz")
    elif i % 5 == 0:
        fizzbuzz.append("Buzz")
    else:
        fizzbuzz.append(str(i))

for fila in range(0, 100, 10):
    print(" ".join(f"{x:>8}" for x in fizzbuzz[fila:fila + 10]))

en_una_linea = ["Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or str(i) for i in range(1, 101)]
tal_cual = sum(x.isdigit() for x in fizzbuzz)
print("\nNúmeros escritos tal cual:", tal_cual)
assert en_una_linea == fizzbuzz
# Inclusión-exclusión: 100 − múltiplos de 3 − múltiplos de 5 + múltiplos de 15
assert tal_cual == 100 - 100 // 3 - 100 // 5 + 100 // 15
```

```python
picos = {"Teide": 3715, "Roque de los Muchachos": 2426, "Pico de las Nieves": 1949, "Garajonay": 1487, "Malpaso": 1501}
for pico, altura in picos.items():
    print(f"{pico:<24}{altura:>6} m  {'par' if altura % 2 == 0 else 'impar'}")
assert [a % 2 == 0 for a in picos.values()] == [False, True, False, False, False]
```

## 2. Filtrar rangos de números

Mostrar separados por comas:

- los números entre 100 y 300 divisibles por 7 pero no por 5;
- los números de 1200 a 2100 cuyas cifras son todas pares (sin convertirlos a cadena);
- los pares entre 10 y 55 salvo el 16 y los múltiplos de 3.

```python
def cifras_pares(n):
    while n:
        if n % 10 % 2:
            return False
        n //= 10
    return True


por_7 = [n for n in range(100, 301) if n % 7 == 0 and n % 5 != 0]
todas_pares = [n for n in range(1200, 2101) if cifras_pares(n)]
pares_filtrados = [n for n in range(10, 56, 2) if n != 16 and n % 3 != 0]

print("Divisibles por 7 y no por 5:", ", ".join(map(str, por_7)))
print("Cifras pares:", ", ".join(map(str, todas_pares)))
print("Pares sin el 16 ni múltiplos de 3:", ", ".join(map(str, pares_filtrados)))

assert todas_pares == [n for n in range(1200, 2101) if all(int(c) % 2 == 0 for c in str(n))]
assert len(por_7) == len(range(105, 301, 7)) - len(range(105, 301, 35))
```

## 3. Series de cifras repetidas

Calcular $a + aa + aaa + aaaa$ para una cifra $a$ (con $a = 7$: $7 + 77 + 777 + 7777$) y, en general, la suma de los $n$ primeros términos $2 + 22 + 222 + \dots$, construyendo cada término a partir del anterior sin convertir a cadena. Comprobarlo con cadenas.

```python
def serie_repetida(cifra, n):
    termino = total = 0
    terminos = []
    for _ in range(n):
        termino = termino * 10 + cifra
        terminos.append(termino)
        total += termino
    return terminos, total


terminos, total = serie_repetida(7, 4)
print(" + ".join(map(str, terminos)), "=", total)
for n in range(1, 7):
    terminos, total = serie_repetida(2, n)
    print(f"n = {n}: {total}")

for cifra in range(1, 10):
    for n in range(1, 10):
        assert serie_repetida(cifra, n)[1] == sum(int(str(cifra) * i) for i in range(1, n + 1))
```

## 4. Números de Armstrong

Un número de $k$ cifras es de Armstrong (o narcisista) si es igual a la suma de sus cifras elevadas a $k$: $153 = 1^3 + 5^3 + 3^3$. Escribir la función sin convertir a cadena y listar los de 3 y 4 cifras.

```python
def es_armstrong(n):
    cifras = []
    m = n
    while m:
        cifras.append(m % 10)
        m //= 10
    k = len(cifras)
    return n == sum(c**k for c in cifras)


tres = [n for n in range(100, 1000) if es_armstrong(n)]
cuatro = [n for n in range(1000, 10000) if es_armstrong(n)]
print("De 3 cifras:", tres)
print("De 4 cifras:", cuatro)
for n in tres + cuatro:
    print(f"{n} = " + " + ".join(f"{c}^{len(str(n))}" for c in str(n)))

assert all(es_armstrong(n) == (n == sum(int(c) ** len(str(n)) for c in str(n))) for n in range(1, 10000))
```

## 5. Capicúas y cifras extremas

Invertir un número con operaciones aritméticas (sin cadenas) para decir si es capicúa, y obtener su cifra mayor y su cifra menor con un bucle. Contar los capicúas menores que 10 000.

```python
def invertir(n):
    invertido = 0
    while n:
        invertido = invertido * 10 + n % 10
        n //= 10
    return invertido


def es_capicua(n):
    return n == invertir(n)


def cifras_extremas(n):
    mayor = menor = n % 10
    n //= 10
    while n:
        mayor, menor = max(mayor, n % 10), min(menor, n % 10)
        n //= 10
    return mayor, menor


for n in [12321, 4510, 38083, 9]:
    mayor, menor = cifras_extremas(n)
    print(f"{n:>6}  invertido {invertir(n):>6}  capicúa: {es_capicua(n)!s:<5}  mayor {mayor}  menor {menor}")

capicuas = sum(es_capicua(n) for n in range(1, 10000))
print("Capicúas entre 1 y 9999:", capicuas)
assert all(es_capicua(n) == (str(n) == str(n)[::-1]) for n in range(1, 10000))
assert all(cifras_extremas(n) == (int(max(str(n))), int(min(str(n)))) for n in range(10000))
assert capicuas == 9 + 9 + 90 + 90
```

## 6. Secuencia de Collatz

Partiendo de $n$, si es par se divide entre 2 y si es impar se cambia por $3n + 1$, hasta llegar a 1. Escribir la secuencia y el número de pasos para 6 y 27, y encontrar el número inicial menor que 100 con la secuencia más larga.

```python
def collatz(n):
    secuencia = [n]
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        secuencia.append(n)
    return secuencia


print("6:", collatz(6))
s27 = collatz(27)
print(f"27: {len(s27) - 1} pasos, valor máximo {max(s27)}")

pasos = {n: len(collatz(n)) - 1 for n in range(1, 100)}
campeon = max(pasos, key=pasos.get)
print(f"Menor que 100 con más pasos: {campeon} ({pasos[campeon]} pasos)")

assert collatz(6) == [6, 3, 10, 5, 16, 8, 4, 2, 1]
assert len(s27) - 1 == 111 and max(s27) == 9232
assert all(len(collatz(2**k)) - 1 == k for k in range(20))
```

## 7. `for`/`else`: buscar un divisor

Buscar el menor divisor (distinto de 1) de varios números probando hasta $\sqrt{n}$. Usar el `else` del `for`, que solo se ejecuta si el bucle no termina con `break`, para avisar de que no se ha encontrado ninguno y, por tanto, el número es primo.

```python
from math import isqrt


def informar_divisor(n):
    for d in range(2, isqrt(n) + 1):
        if n % d == 0:
            print(f"{n} = {d} × {n // d}")
            return d
    else:
        print(f"{n}: ningún divisor hasta {isqrt(n)}, es primo")
        return None


resultados = {n: informar_divisor(n) for n in [91, 97, 221, 223, 1001, 7919]}
for n, d in resultados.items():
    divisores = [k for k in range(2, n) if n % k == 0]
    assert d == (divisores[0] if divisores else None)
```

## 8. Cinco formas de recorrer del 1 al 100

Obtener la lista de los números del 1 al 100 con un `for`, un `while`, una función recursiva, una comprensión y `map` sobre un `range`, y comprobar que las cinco dan lo mismo. Sumarlos con la fórmula de Gauss.

```python
con_for = []
for i in range(1, 101):
    con_for.append(i)

con_while = []
i = 1
while i <= 100:
    con_while.append(i)
    i += 1


def recursiva(desde, hasta):
    return [] if desde > hasta else [desde] + recursiva(desde + 1, hasta)


formas = {
    "for": con_for,
    "while": con_while,
    "recursión": recursiva(1, 100),
    "comprensión": [i for i in range(1, 101)],
    "map/range": list(map(int, range(1, 101))),
}
for nombre, lista in formas.items():
    print(f"{nombre:<12} {lista[:3]} … {lista[-2:]}  suma {sum(lista)}")

assert all(lista == con_for for lista in formas.values())
assert sum(con_for) == 100 * 101 // 2
```

## 9. Triángulo de Pascal

Construir las $n$ primeras filas del triángulo de Pascal sumando las parejas de la fila anterior y mostrarlas centradas. Comprobar que cada número es $\binom{i}{j}$ y que la fila $i$ suma $2^i$.

```python
from math import comb


def pascal(n):
    filas = [[1]]
    for _ in range(n - 1):
        anterior = filas[-1]
        filas.append([1] + [a + b for a, b in zip(anterior, anterior[1:])] + [1])
    return filas


filas = pascal(10)
ancho = len(" ".join(f"{x:>3}" for x in filas[-1]))
for fila in filas:
    print(" ".join(f"{x:>3}" for x in fila).center(ancho).rstrip())

assert all(x == comb(i, j) for i, fila in enumerate(filas) for j, x in enumerate(fila))
assert all(sum(fila) == 2**i for i, fila in enumerate(filas))
```

## 10. Primos gemelos y triples pitagóricos

Listar las parejas de primos gemelos (primos que se diferencian en 2) hasta 200 con una criba de Eratóstenes. Después, encontrar con bucles anidados los triples pitagóricos $a < b < c$ con $a^2 + b^2 = c^2$ y $c \le 50$, y comprobarlos con la fórmula de Euclides: $k(m^2 - n^2)$, $2kmn$, $k(m^2 + n^2)$ con $m > n > 0$.

```python
def criba(limite):
    es_primo = [True] * (limite + 1)
    es_primo[0] = es_primo[1] = False
    for i in range(2, isqrt(limite) + 1):
        if es_primo[i]:
            for j in range(i * i, limite + 1, i):
                es_primo[j] = False
    return es_primo


N = 200
es_primo = criba(N)
gemelos = [(p, p + 2) for p in range(2, N - 1) if es_primo[p] and es_primo[p + 2]]
print(f"{len(gemelos)} parejas de gemelos hasta {N}:", gemelos)
assert all(all(x % d for d in range(2, x)) for par in gemelos for x in par)
```

```python
C = 50
triples = []
for c in range(1, C + 1):
    for b in range(1, c):
        for a in range(1, b):
            if a * a + b * b == c * c:
                triples.append((a, b, c))
print(f"{len(triples)} triples con c ≤ {C}:")
print(triples)

por_euclides = set()
for m in range(2, isqrt(C) + 1):
    for n in range(1, m):
        k = 1
        while k * (m * m + n * n) <= C:
            a, b = sorted((k * (m * m - n * n), 2 * k * m * n))
            por_euclides.add((a, b, k * (m * m + n * n)))
            k += 1
assert set(triples) == por_euclides
```

## 11. Patrones: filas, letras y cuadrado hueco

Escribir funciones que devuelvan como cadena (una línea por fila) estos dibujos de $n$ filas:

- cada fila repite su número tantas veces como indica;
- una pirámide centrada de letras que sube hasta la $n$-ésima y vuelve a bajar (`A`, `ABA`, `ABCBA`…);
- un cuadrado de asteriscos hueco de lado $n$.

```python
from string import ascii_uppercase


def filas_repetidas(n):
    return "\n".join(" ".join([str(i)] * i) for i in range(1, n + 1))


def piramide_letras(n):
    filas = []
    for i in range(n):
        subida = ascii_uppercase[:i + 1]
        filas.append(" " * (n - 1 - i) + subida + subida[-2::-1])
    return "\n".join(filas)


def cuadrado_hueco(n):
    filas = []
    for i in range(n):
        if i in (0, n - 1):
            filas.append("*" * n)
        else:
            filas.append("*" + " " * (n - 2) + "*")
    return "\n".join(filas)


print("\n\n".join([filas_repetidas(5), piramide_letras(5), cuadrado_hueco(5)]))

assert filas_repetidas(3) == "1\n2 2\n3 3 3"
assert piramide_letras(3) == "  A\n ABA\nABCBA"
assert cuadrado_hueco(4) == "****\n*  *\n*  *\n****"
assert all(fila.strip() == fila.strip()[::-1] for fila in piramide_letras(10).splitlines())
```

## 12. Patrones: rombo y escaleras

Escribir `rombo(n)`, con $2n - 1$ filas y la fila central de $2n - 1$ asteriscos, y `escalera(n, sube)`, que dibuje con `_` y `|` una escalera de $n$ peldaños que sube o baja de izquierda a derecha. Sin espacios al final de las líneas.

```python
def rombo(n):
    mitad = [" " * (n - 1 - i) + "*" * (2 * i + 1) for i in range(n)]
    return "\n".join(mitad + mitad[-2::-1])


def escalera(n, sube=True):
    if sube:
        filas = [" " * (2 * n) + "_"] + [" " * (2 * (n - i)) + "_|" for i in range(1, n + 1)]
    else:
        filas = ["_"] + [" " * (2 * i + 1) + "|_" for i in range(n)]
    return "\n".join(filas)


print("\n\n".join([rombo(4), escalera(4), escalera(4, sube=False)]))

assert rombo(2) == " *\n***\n *"
assert sum(fila.count("*") for fila in rombo(6).splitlines()) == 6**2 + 5**2
assert escalera(2) == "    _\n  _|\n_|"
assert escalera(2, sube=False) == "_\n |_\n   |_"
assert all(fila == fila.rstrip() for n in range(1, 8) for fila in escalera(n).splitlines() + rombo(n).splitlines())
```

## 13. Triángulo de triángulos

Dibujar tres triángulos de asteriscos de altura $h$: uno arriba, centrado, y dos debajo, uno junto al otro y separados por un espacio (como la Trifuerza). Comprobar que tiene $3h^2$ asteriscos y que es simétrico.

```python
def trifuerza(h):
    triangulo = ["*" * (2 * i + 1) for i in range(h)]
    arriba = [" " * (h + h - 1 - i) + fila for i, fila in enumerate(triangulo)]
    abajo = [" " * (h - 1 - i) + fila + " " * (2 * (h - 1 - i) + 1) + fila for i, fila in enumerate(triangulo)]
    return "\n".join(arriba + abajo)


print(trifuerza(4))

for h in range(1, 10):
    dibujo = trifuerza(h)
    ancho = 4 * h - 1
    assert dibujo.count("*") == 3 * h * h
    assert all(fila.ljust(ancho) == fila.ljust(ancho)[::-1] for fila in dibujo.splitlines())
```

## 14. Cifras que menguan y crecen

Dibujar `12345678` y, en cada fila, quitar la última cifra hasta quedarse con `1`; después volver a añadirlas una a una hasta recuperar `12345678`.

```python
def menguan_y_crecen(n):
    cifras = "".join(str(i) for i in range(1, n + 1))
    longitudes = list(range(n, 0, -1)) + list(range(2, n + 1))
    return "\n".join(cifras[:k] for k in longitudes)


dibujo = menguan_y_crecen(8)
print(dibujo)
filas = dibujo.splitlines()
assert len(filas) == 2 * 8 - 1 and filas[0] == filas[-1] == "12345678" and filas[7] == "1"
assert filas == filas[::-1]
```

## 15. Binarios divisibles por 5

En una cadena de números binarios de 4 cifras separados por comas, quedarse con los que son divisibles por 5. Convertir cada uno recorriendo sus bits (sin `int(…, 2)`) y comprobarlo con `int`.

```python
binarios = "0101,1110,1010,0011,1111,0000,1001,0110,1011,1100"


def a_decimal(bits):
    valor = 0
    for bit in bits:
        valor = valor * 2 + (bit == "1")
    return valor


divisibles = [b for b in binarios.split(",") if a_decimal(b) % 5 == 0]
print(", ".join(f"{b} ({a_decimal(b)})" for b in divisibles))
assert divisibles == [b for b in binarios.split(",") if int(b, 2) % 5 == 0]
assert all(a_decimal(format(n, "04b")) == n for n in range(16))
```

## 16. Adivinar un número

El ordenador piensa un número del 1 al 100 y responde «mayor» o «menor» a cada intento. Simular una partida con una lista fija de intentos y después escribir un jugador automático que busque partiendo el intervalo por la mitad. Comprobar que acierta cualquier número en 7 intentos como mucho ($2^7 = 128 \ge 100$).

```python
def pista(secreto, intento):
    if intento < secreto:
        return "mayor"
    if intento > secreto:
        return "menor"
    return "¡acertado!"


secreto = 42
for n, intento in enumerate([50, 30, 45, 40, 43, 42, 60], start=1):
    respuesta = pista(secreto, intento)
    print(f"Intento {n}: {intento:>3} → {respuesta}")
    if respuesta == "¡acertado!":
        break
```

```python
def adivinar(secreto, bajo=1, alto=100):
    intentos = []
    while True:
        medio = (bajo + alto) // 2
        intentos.append(medio)
        match pista(secreto, medio):
            case "mayor":
                bajo = medio + 1
            case "menor":
                alto = medio - 1
            case _:
                return intentos


print("Secreto 42:", adivinar(42))
print("Secreto 100:", adivinar(100))
necesarios = {s: len(adivinar(s)) for s in range(1, 101)}
peor = max(necesarios.values())
print("Peor caso:", peor, "intentos; números que lo necesitan:", sum(v == peor for v in necesarios.values()))
assert all(adivinar(s)[-1] == s for s in range(1, 101))
assert peor <= 7
```
