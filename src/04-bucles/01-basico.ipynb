# Bucles · Básico

`for`, `while` y `range`.

## 1. Palabra 10 veces

Pedir una palabra y mostrarla 10 veces.

```python
palabra = input("Palabra: ")
for _ in range(10):
    print(palabra)
```

## 2. Años cumplidos

Pedir la edad y mostrar todos los años cumplidos, de 1 a la edad.

```python
edad = int(input("Edad: "))
for año in range(1, edad + 1):
    print(año)
```

## 3. Impares hasta n

Pedir un entero positivo y mostrar los impares desde 1 hasta él separados por comas.

```python
n = int(input("Entero positivo: "))
print(", ".join(str(i) for i in range(1, n + 1, 2)))
```

## 4. Cuenta atrás

Pedir un entero positivo y mostrar la cuenta atrás hasta 0 separada por comas.

```python
n = int(input("Entero positivo: "))
print(", ".join(str(i) for i in range(n, -1, -1)))
```

## 5. Múltiplos de n

Pedir $n$ y $m$ y mostrar los múltiplos de $n$ entre $n$ y $m \cdot n$, ambos incluidos.

```python
n = int(input("n: "))
m = int(input("m: "))
for i in range(n, m * n + 1, n):
    print(i)
```

## 6. Potencias de 2

Mostrar las potencias de 2 entre $2^0$ y $2^{30}$, ambas incluidas.

```python
for i in range(31):
    print(2**i)
```

## 7. Sumatorio entre n y m

Pedir dos enteros $n$ y $m$ y mostrar $\sum_{i=n}^{m} i$ con un bucle `for`.

```python
n = int(input("n: "))
m = int(input("m: "))
total = 0
for i in range(n, m + 1):
    total += i
print(total)
```

## 8. Suma de cuadrados

Pedir $n$ y $m$ y mostrar $\sum_{i=n}^{m} i^2$.

```python
n = int(input("n: "))
m = int(input("m: "))
print(sum(i**2 for i in range(n, m + 1)))
```

## 9. Suma de pares en un intervalo

Pedir $n$ y $m$ y sumar los números pares comprendidos entre ellos (incluidos si son pares).

```python
n = int(input("n: "))
m = int(input("m: "))
# primer par >= n
print(sum(range(n + n % 2, m + 1, 2)))
```

## 10. Leer número en rango

Pedir un número entre 0 y 10 (incluidos) y volver a pedirlo mientras esté fuera de rango.

```python
n = float(input("Número (0-10): "))
while not 0 <= n <= 10:
    n = float(input("Fuera de rango. Número (0-10): "))
print("Válido:", n)
```

## 11. Contraseña hasta acertar

Guardar una contraseña y pedirla al usuario hasta que la introduzca correctamente.

```python
contraseña = "secreto"
while input("Contraseña: ") != contraseña:
    print("Incorrecta")
print("Acceso concedido")
```

## 12. Eco hasta "salir"

Repetir todo lo que escriba el usuario hasta que escriba `salir`.

```python
while (texto := input("> ")) != "salir":
    print(texto)
```

## 13. De 0 a 10 y de 10 a 0

Recorrer de 0 a 10 y de 10 a 0, primero con `for` y luego con `while`.

```python
print(*range(11))
print(*range(10, -1, -1))
i = 0
while i <= 10:
    print(i, end=" ")
    i += 1
print()
while i > 0:
    i -= 1
    print(i, end=" ")
print()
```

## 14. Pares e impares hasta 100

Recorrer de 0 a 100 mostrando solo los pares y después solo los impares.

```python
print([i for i in range(101) if i % 2 == 0])
print([i for i in range(101) if i % 2])
```

## 15. Sumas de 0 a 100

Sumar con un bucle `for` todos los números de 0 a 100, y por separado los pares y los impares.

```python
total = pares = impares = 0
for i in range(101):
    total += i
    if i % 2 == 0:
        pares += i
    else:
        impares += i
print(f"La suma de todos es {total}. Pares: {pares}. Impares: {impares}.")
```
