## 1. Trazas de bucles `while`

Hacer la traza de estos bucles: `i` de 0 mientras `i <= 3` sumando 1; de 0 mientras `i < 10` sumando 2; de 3 mientras `i < 10` sumando 2 antes de imprimir; de 1 mientras `i < 100` duplicando; y de 10 mientras `i < 2` duplicando (no entra nunca).

```python
def traza(inicio, condicion, paso, imprimir_antes=True):
    i, salida = inicio, []
    while condicion(i):
        if imprimir_antes:
            salida.append(i)
        i = paso(i)
        if not imprimir_antes:
            salida.append(i)
    return salida

print(traza(0, lambda i: i <= 3, lambda i: i + 1))
print(traza(0, lambda i: i < 10, lambda i: i + 2))
print(traza(3, lambda i: i < 10, lambda i: i + 2, False))
print(traza(1, lambda i: i < 100, lambda i: i * 2, False))
print(traza(10, lambda i: i < 2, lambda i: i * 2, False))
```

## 2. Trazas con valores leídos

Probar el bucle `while i < limite: print(i); i += incremento` con distintos valores de `i`, `limite` e `incremento`. ¿Qué pasa si `i >= limite`? ¿Y si el incremento es 0 o negativo?

Si `i >= limite` no se ejecuta ninguna vez; con incremento 0 o negativo (y `i < limite`) el bucle no termina nunca.

```python
for i, limite, incremento in ((5, 10, 1), (12, 10, 1), (-3, 2, 2), (0, 10, 3)):
    valores = []
    while i < limite:
        valores.append(i)
        i += incremento
    print(valores)
```

## 4. Dos formas de sumar

¿Producen el mismo resultado sumar de 1 a 1000 incrementando `i` antes o después de acumular?

```python
s1, i = 0, 1
while i <= 1000:
    s1 += i
    i += 1
s2, i = 0, 0
while i < 1000:
    i += 1
    s2 += i
print(s1, s2, s1 == s2)
```

## 6. Factorial y combinaciones

Calcular $n!$ con un bucle y $C_n^m = \frac{n!}{(n-m)!\,m!}$ (con $n=15, m=10$ es 3003).

```python
def factorial(n):
    f = 1
    for i in range(2, n + 1):
        f *= i
    return f

n = int(input("n: "))
m = int(input("m: "))
if n >= m:
    print(factorial(n) // (factorial(n - m) * factorial(m)))
```

## 8. Menú del círculo repetido

Mostrar el menú (diámetro, perímetro, área) hasta elegir una opción válida y, en otra versión, pedir el radio en cada cálculo hasta elegir salir.

```python
from math import pi

radio = float(input("Radio: "))
opcion = ""
while opcion not in ("a", "b", "c"):
    opcion = input("a) diámetro b) perímetro c) área: ")
    if opcion not in ("a", "b", "c"):
        print("Solo hay tres opciones: a, b o c. Has tecleado", opcion)
print({"a": 2 * radio, "b": 2 * pi * radio, "c": pi * radio**2}[opcion])

while (radio := input("Radio (s para salir): ")) != "s":
    radio = float(radio)
    opcion = input("a) diámetro b) perímetro c) área: ")
    print({"a": 2 * radio, "b": 2 * pi * radio, "c": pi * radio**2}.get(opcion, "Opción no válida"))
```

## 9. Calculadora de vectores 3D

Menú para leer dos vectores y calcular suma, diferencia, producto escalar, producto vectorial, ángulo en grados y longitudes, hasta elegir finalizar. Si se elige una opción no válida, avisar.

```python
from math import acos, degrees, sqrt

v1 = v2 = (0, 0, 0)
while (opcion := input("1-2 leer, 3 suma, 4 resta, 5 escalar, 6 vectorial, 7 ángulo, 8 longitudes, 9 fin: ")) != "9":
    if opcion in ("1", "2"):
        v = tuple(float(x) for x in input("x y z: ").split())
        v1, v2 = (v, v2) if opcion == "1" else (v1, v)
    elif opcion == "3":
        print(tuple(a + b for a, b in zip(v1, v2)))
    elif opcion == "4":
        print(tuple(a - b for a, b in zip(v1, v2)))
    elif opcion == "5":
        print(sum(a * b for a, b in zip(v1, v2)))
    elif opcion == "6":
        (x1, y1, z1), (x2, y2, z2) = v1, v2
        print((y1 * z2 - z1 * y2, z1 * x2 - x1 * z2, x1 * y2 - y1 * x2))
    elif opcion == "7":
        escalar = sum(a * b for a, b in zip(v1, v2))
        print(degrees(acos(escalar / (sqrt(sum(a * a for a in v1)) * sqrt(sum(b * b for b in v2))))))
    elif opcion == "8":
        print(sqrt(sum(a * a for a in v1)), sqrt(sum(b * b for b in v2)))
    else:
        print("Opción no válida")
```

## 12. Trazas del test de primalidad

Trazar el test de primalidad (probar divisores de 2 a $n-1$) para 4, 13, 18 y 2 (¡con 2 no entra en el bucle!), y para 125 con las versiones que se detienen con `break`.

```python
def traza_primo(n, con_break=False):
    divisores_probados, es_primo = [], True
    for d in range(2, n):
        divisores_probados.append(d)
        if n % d == 0:
            es_primo = False
            if con_break:
                break
    return es_primo, divisores_probados

for n in (4, 13, 18, 2):
    primo, probados = traza_primo(n)
    print(n, primo, probados)
print(125, traza_primo(125, con_break=True))
print(125, len(traza_primo(125)[1]), "divisores probados sin break")
```

## 13. Menos divisores para saber si es primo

Basta probar divisores hasta $\lfloor n/2 \rfloor$ (ningún divisor propio supera $n/2$) y, mejor aún, hasta $\lfloor\sqrt{n}\rfloor$ (si $n = a \cdot b$, uno de los dos es $\le \sqrt n$).

```python
from math import isqrt

def primo_mitad(n):
    return n > 1 and all(n % d for d in range(2, n // 2 + 1))

def primo_raiz(n):
    return n > 1 and all(n % d for d in range(2, isqrt(n) + 1))

assert [primo_mitad(n) for n in range(200)] == [primo_raiz(n) for n in range(200)]
print([n for n in range(50) if primo_raiz(n)])
```
