## 1. Suma de números binarios

Leer dos cadenas binarias y mostrar su suma como cadena binaria, operando con acarreo directamente sobre los caracteres. Ejemplo: `'100' + '111' = '1011'`.

```python
a = input("Primer binario: ")
b = input("Segundo binario: ")
a, b = a.zfill(len(b)), b.zfill(len(a))
resultado, acarreo = "", 0
for x, y in zip(reversed(a), reversed(b)):
    acarreo, bit = divmod(int(x) + int(y) + acarreo, 2)
    resultado = str(bit) + resultado
print(("1" if acarreo else "") + resultado)
```

## 2. Cifrado César

Sustituir cada letra (mayúscula o minúscula) y cada dígito por el que está $n$ posiciones a la derecha, dando la vuelta al final (con $n=2$: `y→a`, `z→b`, `9→1`). Leer un texto y $n$ y mostrar el texto cifrado.

```python
from string import ascii_lowercase, ascii_uppercase, digits

def cesar(texto, n):
    resultado = ""
    for c in texto:
        for alfabeto in (ascii_lowercase, ascii_uppercase, digits):
            if c in alfabeto:
                c = alfabeto[(alfabeto.index(c) + n) % len(alfabeto)]
                break
        resultado += c
    return resultado

texto = input("Texto: ")
n = int(input("n: "))
print(cesar(texto, n))
```

## 3. Descifrado César

Leer un texto cifrado con la técnica anterior y el valor de $n$, y mostrar el texto original.

```python
# descifrar = cifrar desplazando hacia atrás
texto = input("Texto cifrado: ")
n = int(input("n: "))
print(cesar(texto, -n))
```

## 4. Prefijos

Mostrar todos los prefijos de una cadena. Ejemplo: `'UJI'` → `U`, `UJ`, `UJI`.

```python
cadena = input("Cadena: ")
for i in range(1, len(cadena) + 1):
    print(cadena[:i])
```

## 5. Subcadenas de longitud k

Leer una cadena y un entero $k$ y mostrar todas sus subcadenas de longitud $k$.

```python
cadena = input("Cadena: ")
k = int(input("k: "))
for i in range(len(cadena) - k + 1):
    print(cadena[i:i + k])
```

## 6. ¿Es subcadena?

Leer dos cadenas `a` y `b` y decir si `b` es subcadena de `a`, sin usar el operador `in`. Ejemplo: `'de'` es subcadena de `'subcadena'`.

```python
a = input("a: ")
b = input("b: ")
print(any(a[i:i + len(b)] == b for i in range(len(a) - len(b) + 1)))
```

## 7. Prefijo común más largo

Leer dos cadenas y mostrar su prefijo común más largo. Ejemplo: `'politécnico'` y `'polinización'` → `'poli'`.

```python
a = input("Primera: ")
b = input("Segunda: ")
i = 0
while i < min(len(a), len(b)) and a[i] == b[i]:
    i += 1
print(a[:i])
```

## 8. Validar correo con expresiones regulares

Escribir una función que valide un correo con `re`. Variantes: solo dominios `.com`, y además rechazar dominios `@google`.

```python
import re

def correo_valido(correo):
    return re.fullmatch(r"[\w.+-]+@[\w-]+(\.[\w-]+)+", correo) is not None

def correo_com(correo):
    return re.fullmatch(r"[\w.+-]+@[\w-]+(\.[\w-]+)*\.com", correo) is not None

def correo_com_sin_google(correo):
    return correo_com(correo) and not re.search(r"@google\.", correo)

for c in ["nombre@dominio.es", "esto@no.es@un.correo", "ana@empresa.com", "bob@google.com"]:
    print(f"{c:22}", correo_valido(c), correo_com(c), correo_com_sin_google(c))
```

## 9. Compresión RLE

Comprimir una secuencia agrupando repeticiones consecutivas (`AAAAETTTTH` → `A4E1T4H1`) devolviendo un `str`, y escribir `descomprime`, que haga lo contrario.

```python
import re
from itertools import groupby

def comprime(s):
    return "".join(f"{c}{len(list(g))}" for c, g in groupby(s))

def descomprime(s):
    return "".join(c * int(n) for c, n in re.findall(r"(\D)(\d+)", s))

comprimido = comprime("AAAAETTTTH")
print(comprimido, descomprime(comprimido))
assert descomprime(comprime("aaabccccccccccccd")) == "aaabccccccccccccd"
```

## 10. ADN complementario

La adenina (A) se empareja con la timina (T) y la citosina (C) con la guanina (G). Escribir `complement`, que devuelva la secuencia complementaria de una cadena de ADN sin usar bibliotecas externas.

```python
def complement(adn):
    return adn.translate(str.maketrans("ATCG", "TAGC"))

print(complement("ATTGCCGATA"))
```
