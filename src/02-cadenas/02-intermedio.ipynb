# Cadenas · Intermedio

`split`, `join`, formato de cadenas, recorrido carácter a carácter y validación.

## 1. Fecha de nacimiento

Pedir la fecha de nacimiento en formato `dd/mm/aaaa` y mostrar día, mes y año. Debe funcionar también si el día o el mes tienen un solo carácter.

```python
dia, mes, año = input("Fecha (dd/mm/aaaa): ").split("/")
print("Día", dia)
print("Mes", mes)
print("Año", año)
```

## 2. Cesta de la compra

Pedir los productos de una cesta separados por comas y mostrar cada uno en una línea.

```python
productos = input("Productos (separados por comas): ")
for producto in productos.split(","):
    print(producto.strip())
```

## 3. Ticket con formato

Pedir nombre de producto, precio unitario y unidades. Mostrar: nombre, precio con 6 dígitos enteros y 2 decimales, unidades con 3 dígitos y total con 8 dígitos enteros y 2 decimales.

```python
nombre = input("Producto: ")
precio = float(input("Precio unitario: "))
unidades = int(input("Unidades: "))
# ancho = enteros + punto + decimales
print(f"{nombre} {precio:09.2f} {unidades:03d} {precio * unidades:011.2f}")
```

## 4. Palabras de longitud k

Leer una frase y un entero $k$ y mostrar cuántas palabras tienen exactamente $k$ caracteres.

```python
palabras = input("Frase: ").split()
k = int(input("k: "))
print(sum(len(p) == k for p in palabras))
```

## 5. ¿Todas las palabras miden k?

Leer una frase y un entero $k$ y decir si todas sus palabras miden $k$ caracteres.

```python
palabras = input("Frase: ").split()
k = int(input("k: "))
print(all(len(p) == k for p in palabras))
```

## 6. Contar dígitos

Mostrar cuántos dígitos tiene una cadena. Ejemplo: `'un 1 y un 20'` tiene 3.

```python
cadena = input("Cadena: ")
print(sum(c.isdigit() for c in cadena))
```

## 7. Contar números

Mostrar cuántos números (secuencias de dígitos seguidos) hay en una cadena. Ejemplo: `'un 1, un 201 y 2 unos'` tiene 3.

```python
cadena = input("Cadena: ")
numeros, anterior = 0, False
for c in cadena:
    # cuenta solo inicio de cada secuencia de dígitos
    if c.isdigit() and not anterior:
        numeros += 1
    anterior = c.isdigit()
print(numeros)
```

## 8. ¿Es un entero bien formado?

Indicar si una cadena es un número entero bien formado (`'12'` sí; `'1 2'` o `'a'` no). Se admite un signo inicial.

```python
cadena = input("Cadena: ")
cifras = cadena[1:] if cadena[:1] in "+-" else cadena
print("Es entero" if cifras.isdecimal() else "No es entero")
```

## 9. ¿Es un identificador válido?

Indicar si una cadena es un identificador de variable válido: empieza por letra o `_` y sigue con letras, dígitos o `_`.

```python
cadena = input("Cadena: ")
valido = cadena != "" and (cadena[0].isalpha() or cadena[0] == "_") and all(c.isalnum() or c == "_" for c in cadena)
print("Identificador válido" if valido else "Identificador inválido")
```

## 10. ¿Es un flotante bien formado?

Indicar si una cadena es un número flotante bien formado. Deben aceptarse `'3.1'`, `'3.'`, `'.1'`, `'1e+5'`, `'-10.2E3'`, `'3.1e-2'` y `'.1e01'`.

```python
import re

FLOTANTE = re.compile(r"[+-]?(\d+\.?\d*|\.\d+)([eE][+-]?\d+)?")

for cadena in ["3.1", "3.", ".1", "1e+5", "-10.2E3", "3.1e-2", ".1e01", "1.2.3", "e5", "."]:
    print(f"{cadena!r:10}", bool(FLOTANTE.fullmatch(cadena)))
```

## 11. Leer un número binario

Pedir una cadena binaria. Si tiene algún carácter distinto de `0` o `1`, avisar y volver a pedirla.

```python
binario = input("Número binario: ")
while not binario or set(binario) - {"0", "1"}:
    print("No es un número binario")
    binario = input("Número binario: ")
print("Leído:", binario)
```

## 12. De octal a decimal

Convertir una cadena de dígitos 0-7 al valor que representa en base 8, sin usar `int(cadena, 8)`.

```python
octal = input("Número octal: ")
valor = 0
for c in octal:
    valor = valor * 8 + int(c)
print(valor)
```

## 13. De hexadecimal a decimal

Convertir una cadena de dígitos y letras a-f al valor que representa en base 16, sin usar `int(cadena, 16)`.

```python
hexa = input("Número hexadecimal: ").lower()
valor = 0
for c in hexa:
    valor = valor * 16 + "0123456789abcdef".index(c)
print(valor)
```
