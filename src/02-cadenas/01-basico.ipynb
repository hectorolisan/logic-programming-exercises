# Cadenas · Básico

Concatenación, repetición, métodos de `str` y slicing.

## 1. Nombre repetido

Pedir un nombre y un entero $n$ y mostrar el nombre $n$ veces, cada vez en una línea.

```python
nombre = input("Nombre: ")
n = int(input("Veces: "))
print("\n".join([nombre] * n))
```

## 2. Mayúsculas y minúsculas

Pedir el nombre completo y mostrarlo en minúsculas, en mayúsculas y con la inicial de cada palabra en mayúscula.

```python
nombre = input("Nombre completo: ")
print(nombre.lower())
print(nombre.upper())
print(nombre.title())
```

## 3. Longitud del nombre

Pedir el nombre y mostrar `<NOMBRE> tiene <n> letras`.

```python
nombre = input("Nombre: ")
print(f"{nombre.upper()} tiene {len(nombre)} letras")
```

## 4. Teléfono sin prefijo ni extensión

Los teléfonos tienen el formato `+34-913724710-56`. Pedir uno y mostrar solo el número central.

```python
telefono = input("Teléfono (+xx-xxxxxxxxx-xx): ")
print(telefono.split("-")[1])
```

## 5. Frase invertida

Pedir una frase y mostrarla al revés.

```python
frase = input("Frase: ")
print(frase[::-1])
```

## 6. Vocal en mayúscula

Pedir una frase y una vocal, y mostrar la frase con esa vocal en mayúscula.

```python
frase = input("Frase: ")
vocal = input("Vocal: ")
print(frase.replace(vocal, vocal.upper()))
```

## 7. Cambio de dominio de correo

Pedir un correo y mostrar otro con el mismo nombre de usuario pero dominio `ceu.es`.

```python
correo = input("Correo: ")
print(correo.split("@")[0] + "@ceu.es")
```

## 8. Euros y céntimos

Pedir un precio con dos decimales y mostrar los euros y los céntimos.

```python
euros, centimos = input("Precio: ").split(".")
print(f"{euros} euros y {centimos} céntimos")
```

## 9. Contar espacios

Leer una cadena y mostrar cuántos espacios en blanco contiene.

```python
cadena = input("Cadena: ")
print(cadena.count(" "))
```

## 10. Contar mayúsculas

Leer una cadena y mostrar cuántas letras mayúsculas contiene.

```python
cadena = input("Cadena: ")
print(sum(c.isupper() for c in cadena))
```

## 11. ¿Contiene algún dígito?

Leer una cadena y mostrar `Contiene dígito` o `No contiene dígito`.

```python
cadena = input("Cadena: ")
print("Contiene dígito" if any(c.isdigit() for c in cadena) else "No contiene dígito")
```

## 12. Métodos de cadenas

Sobre `'Coding For All'`: longitud, mayúsculas, minúsculas, `capitalize`, `title`, `swapcase`, primera palabra, si contiene `Coding`, reemplazar `Coding` por `Python`, separar por espacios, carácter en el índice 0, último índice, carácter en el índice 10, si empieza por `Coding` y si termina por `coding`. Además: cambiar *Python for Everyone* por *Python for All*, separar una lista de empresas por comas, quitar espacios a los extremos y comprobar identificadores.

```python
company = "Coding For All"
print(len(company), company.upper(), company.lower())
print(company.capitalize(), company.title(), company.swapcase())
print(company.split()[0], "Coding" in company, company.replace("Coding", "Python"))
print(company.split(), company[0], len(company) - 1, company[10])
print(company.startswith("Coding"), company.endswith("coding"))
print("Python for Everyone".replace("Everyone", "All"))
print("Facebook, Google, Microsoft, Apple, IBM, Oracle, Amazon".split(", "))
print(repr("   Coding For All      ".strip()))
print("30DaysOfPython".isidentifier(), "thirty_days_of_python".isidentifier())
```

## 13. Acrónimos

Crear el acrónimo de `'Python For Everyone'` y de `'Coding For All'`.

```python
def acronimo(texto):
    return "".join(palabra[0] for palabra in texto.split()).upper()

print(acronimo("Python For Everyone"), acronimo("Coding For All"))
```

## 14. Buscar y recortar

En *You cannot end a sentence with because because because is a conjunction*, encontrar la primera y la última aparición de `because` y recortar la frase `'because because because'`. En `'Coding For All'`, la primera `C` y la primera `F`; en `'Coding For All People'`, la última `l`.

```python
frase = "You cannot end a sentence with because because because is a conjunction"
inicio = frase.index("because")
fin = frase.rindex("because") + len("because")
print(inicio, frase.rindex("because"), repr(frase[inicio:fin]))
print("Coding For All".index("C"), "Coding For All".index("F"), "Coding For All People".rfind("l"))
```

## 15. Formato y secuencias de escape

Unir `['Django', 'Flask', 'Bottle', 'Pyramid', 'Falcon']` con `' # '`, escribir dos frases en líneas distintas, una tabla con tabuladores y las operaciones entre 8 y 6 con formato.

```python
print(" # ".join(["Django", "Flask", "Bottle", "Pyramid", "Falcon"]))
print("I am enjoying this challenge.\nI just wonder what is next.")
print("Name\tAge\tCountry\tCity\nAsabeneh\t250\tFinland\tHelsinki")
radius = 10
print(f"The area of a circle with radius {radius} is {3.14 * radius**2:.0f} meters square.")
a, b = 8, 6
for op, r in (("+", a + b), ("-", a - b), ("*", a * b), ("/", a / b), ("%", a % b), ("//", a // b), ("**", a**b)):
    print(f"{a} {op} {b} = {round(r, 2)}")
```
