## 5. Contar palabras por transiciones

Contar palabras contando transiciones de carácter no blanco a blanco, trazar el programa con `'a b'` y `'a  b'`, comprobar qué ocurre con una cadena de solo espacios y contar en su lugar transiciones de blanco a no blanco. ¿Es correcta la propuesta del aprendiz que resta 1 si la cadena acaba en espacio?

La versión basada en transiciones de blanco a no blanco funciona siempre; la del aprendiz falla con cadenas que empiezan o están formadas solo por espacios.

```python
def palabras_no_blanco_a_blanco(cadena, traza=False):
    cambios, anterior = 0, " "
    for caracter in cadena + " ":
        if caracter == " " and anterior != " ":
            cambios += 1
        if traza:
            print(f"  carácter={caracter!r} anterior={anterior!r} cambios={cambios}")
        anterior = caracter
    return cambios

def palabras_blanco_a_no_blanco(cadena):
    cambios, anterior = 0, " "
    for caracter in cadena:
        if caracter != " " and anterior == " ":
            cambios += 1
        anterior = caracter
    return cambios

def palabras_aprendiz(cadena):
    cambios = 0
    for i in range(1, len(cadena)):
        if cadena[i] == " " and cadena[i - 1] != " ":
            cambios += 1
    if cadena[-1] == " ":
        cambios -= 1
    return cambios + 1

palabras_no_blanco_a_blanco("a b", traza=True)
for c in ("a b", "a  b", "   ", "  hola  mundo ", "uno"):
    print(repr(c), palabras_no_blanco_a_blanco(c), palabras_blanco_a_no_blanco(c), palabras_aprendiz(c), len(c.split()))
```

## 6. Palabras largas y cortas

Leer una frase y un entero $k$ y decir: si alguna palabra mide $k$; `Hay palabras largas` si alguna mide $k$ o más; `Todas son cortas` si todas miden menos de $k$; y `Todas las palabras son largas` si todas miden $k$ o más.

```python
palabras = input("Frase: ").split()
k = int(input("k: "))
print("Alguna mide k:", any(len(p) == k for p in palabras))
print("Hay palabras largas" if any(len(p) >= k for p in palabras) else "No hay palabras largas")
print("Todas son cortas" if all(len(p) < k for p in palabras) else "Hay alguna palabra larga")
print("Todas las palabras son largas" if all(len(p) >= k for p in palabras) else "Hay alguna palabra corta")
```

## 7. Paréntesis equilibrados

Decir si una cadena está bien parentizada: cada `(` debe cerrarse después con `)`.

```python
def bien_parentizada(texto):
    abiertos = 0
    for c in texto:
        abiertos += {"(": 1, ")": -1}.get(c, 0)
        if abiertos < 0:
            return False
    return abiertos == 0

for t in ("Esto (es (un) (ejemplo (de) ((cadena) bien)) parentizada).", "una cadena)", "(una cadena", "(una (cadena)", ")una( cadena"):
    print(bien_parentizada(t), t)
```

## 8. Variantes de binario a decimal

Trazar con `'1101'` y `'010'` tres variantes: `valor = 2 * valor + 1` / `valor = 2 * valor`; `valor += valor + 1` / `valor += valor`; y `valor += valor + int(bit)`. ¿Son correctas? Después, pedir de nuevo la cadena si no es binaria.

Las tres son correctas: todas calculan $v \leftarrow 2v + b$.

```python
def variante1(bits):
    valor = 0
    for bit in bits:
        valor = 2 * valor + 1 if bit == "1" else 2 * valor
    return valor

def variante2(bits):
    valor = 0
    for bit in bits:
        valor += valor + 1 if bit == "1" else valor
    return valor

def variante3(bits):
    valor = 0
    for bit in bits:
        valor += valor + int(bit)
    return valor

for b in ("1101", "010"):
    print(b, variante1(b), variante2(b), variante3(b))

bits = input("Número binario: ")
while not bits or set(bits) - {"0", "1"}:
    bits = input("Número binario mal formado. Otra vez: ")
print(variante3(bits))
```

## 9. Octal, decimal o hexadecimal

Interpretar una cadena: si empieza por `0x`/`0X` es hexadecimal (`'0xff'` = 255), si empieza por `0` es octal (`'017'` = 15) y si no, decimal.

```python
def valor(cadena):
    if cadena[:2].lower() == "0x":
        return int(cadena[2:], 16)
    if cadena.startswith("0") and len(cadena) > 1:
        return int(cadena[1:], 8)
    return int(cadena)

print(valor("0xff"), valor("017"), valor("99"), valor("0"))
```

## 11. Palabra alfabética

Una palabra es alfabética si sus letras están en orden alfabético (amor, chino, himno).

```python
def es_alfabetica(palabra):
    return all(a <= b for a, b in zip(palabra, palabra[1:]))

print([(p, es_alfabetica(p)) for p in ("amor", "chino", "himno", "perro")])
```

## 12. Frases palíndromas

Decir si una cadena es palíndromo; después, si una frase lo es ignorando espacios y signos; y, por último, también ignorando mayúsculas, tildes y diéresis (*Dábale arroz a la zorra el abad*).

```python
import unicodedata

def es_palindromo(cadena):
    return cadena == cadena[::-1]

def normaliza(frase):
    # NFD separa la letra de su tilde
    sin_tildes = "".join(c for c in unicodedata.normalize("NFD", frase.lower()) if unicodedata.category(c) != "Mn")
    return "".join(c for c in sin_tildes if c.isalnum())

frases = ["ana", "anita lava la tina", "luz azul", "Dábale arroz a la zorra el abad", "Amigo, no gima",
          "A mamá Roma le aviva el amor a papá, y a papá Roma le aviva el amor a mamá", "¡arriba la birra!", "no lo es"]
for f in frases:
    print(es_palindromo(normaliza(f)), f)
```

## 17. Texto con secuencias `=XX`

Un programa decodifica textos donde `=XX` representa un carácter en hexadecimal (como el formato *quoted-printable* del correo). Con `'Hola =A. ='` falla porque `=A.` no son dos cifras hexadecimales y el último `=` no va seguido de nada. Corregirlo dejando intactas las secuencias inválidas.

```python
import string

def decodifica(texto):
    resultado, i = "", 0
    while i < len(texto):
        codigo = texto[i + 1:i + 3]
        if texto[i] == "=" and len(codigo) == 2 and all(c in string.hexdigits for c in codigo):
            resultado += chr(int(codigo, 16))
            i += 3
        else:
            resultado += texto[i]
            i += 1
    return resultado

print(decodifica("Hola =41 y =3D"), "|", decodifica("Hola =A. ="))
```

## 18. Cadenas en memoria

Explicar el estado de la memoria tras `a = 'cadena'; b = a[2:3]; c = b + ''` y tras `a = 'ab'; a *= 3; b = a; c = a[:]; c = c + b`. ¿Qué se muestra al final?

Las cadenas son inmutables: `b = a` hace que ambas variables apunten al mismo objeto, y cada operación crea objetos nuevos.

```python
a = "cadena"
b = a[2:3]
c = b + ""
print(a, b, c, b is c)
a = "ab"
a *= 3
b = a
c = a[:]
print(a is b, a is c)
c = c + b
print(a, b, c)
```
