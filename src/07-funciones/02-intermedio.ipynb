## 1. Estadísticos

Escribir una función que reciba una lista de números y devuelva un diccionario con su media, varianza y desviación típica.

```python
def estadisticos(muestra):
    media = sum(muestra) / len(muestra)
    varianza = sum((x - media) ** 2 for x in muestra) / len(muestra)
    return {"media": media, "varianza": varianza, "desviación típica": varianza**0.5}

print(estadisticos([2, 4, 4, 4, 5, 5, 7, 9]))
```

## 2. MCD y mcm

Escribir una función que calcule el máximo común divisor de dos números (algoritmo de Euclides) y otra que calcule el mínimo común múltiplo.

```python
def mcd(a, b):
    while b:
        a, b = b, a % b
    return a

def mcm(a, b):
    return a * b // mcd(a, b)

print(mcd(84, 36), mcm(4, 6))
```

## 3. Binario y decimal

Escribir una función que convierta un decimal a binario y otra que convierta un binario a decimal, sin usar `bin` ni `int(x, 2)`.

```python
def decimal_a_binario(n):
    binario = ""
    while n > 1:
        n, bit = divmod(n, 2)
        binario = str(bit) + binario
    return str(n) + binario

def binario_a_decimal(b):
    return sum(int(bit) * 2**i for i, bit in enumerate(reversed(b)))

print(decimal_a_binario(27), binario_a_decimal("11011"))
```

## 4. Frecuencia de palabras

Escribir una función que reciba un texto y devuelva un diccionario con cada palabra y su frecuencia, y otra que reciba ese diccionario y devuelva una tupla con la palabra más repetida y su frecuencia.

```python
def frecuencias(texto):
    cuenta = {}
    for palabra in texto.lower().split():
        cuenta[palabra] = cuenta.get(palabra, 0) + 1
    return cuenta

def mas_repetida(cuenta):
    palabra = max(cuenta, key=cuenta.get)
    return palabra, cuenta[palabra]

f = frecuencias("Como quieres que te quiera si el que quiero que me quiera no me quiere como quiero que me quiera")
print(mas_repetida(f))
```

## 5. ¿Es repetición?

Escribir `es_repeticion`, que diga si una cadena es otra cadena concatenada consigo misma (`'abab'` → `True`, `'ababab'` → `False`).

```python
def es_repeticion(s):
    mitad = len(s) // 2
    return len(s) % 2 == 0 and s[:mitad] == s[mitad:]

print(es_repeticion("abab"), es_repeticion("ababab"))
```

## 6. Números perfectos

Un número es perfecto si es igual a la suma de sus divisores propios (6 = 1 + 2 + 3). Escribir una función que devuelva los perfectos entre 1 y $n$.

```python
def es_perfecto(n):
    return n == sum(d for d in range(1, n // 2 + 1) if n % d == 0)

def perfectos(n):
    return [i for i in range(2, n + 1) if es_perfecto(i)]

print(perfectos(10000))
```

## 7. Suma de diferencias contiguas

Escribir una función que sume las diferencias entre elementos contiguos de una lista (`[1, 3, 6, 10]` → 2 + 3 + 4 = 9). ¿Se puede hacer sin bucles ni `sum`?

```python
def suma_diferencias(lista):
    # la suma telescópica se reduce a último - primero
    return lista[-1] - lista[0] if lista else 0

print(suma_diferencias([1, 3, 6, 10]))
```

## 8. Cadenas más largas

Escribir una función que devuelva la cadena más larga de una lista y otra que devuelva todas las que empatan como más largas.

```python
def mas_larga(cadenas):
    return max(cadenas, key=len)

def todas_mas_largas(cadenas):
    n = len(mas_larga(cadenas))
    return [c for c in cadenas if len(c) == n]

print(mas_larga(["Pepe", "Juan", "María", "Ana"]))
print(todas_mas_largas(["Pepe", "Ana", "Juan", "Paz"]))
```

## 9. Ecuación lineal

Escribir una función que resuelva $ax + b = 0$ y devuelva `None` si no tiene solución o tiene infinitas.

```python
def ecuacion_lineal(a, b):
    return -b / a if a != 0 else None

print(ecuacion_lineal(2, -8), ecuacion_lineal(0, 3))
```

## 10. Letra del DNI

La letra del DNI se obtiene del resto de dividir el número entre 23, usando la tabla `TRWAGMYFPDXBNJZSQVHLCKE`. Escribir `letra_dni` y `comprueba_letra_dni`, que diga si la letra corresponde al número.

```python
def letra_dni(numero):
    return "TRWAGMYFPDXBNJZSQVHLCKE"[numero % 23]

def comprueba_letra_dni(numero, letra):
    return letra_dni(numero) == letra.upper()

print(letra_dni(12345678), comprueba_letra_dni(12345678, "z"), comprueba_letra_dni(12345678, "A"))
```

## 11. Mínimo y máximo a la vez

Escribir una función que devuelva simultáneamente el mínimo y el máximo de una lista de enteros, y otra que devuelva la primera y la última palabra en orden alfabético.

```python
def min_max(lista):
    return min(lista), max(lista)

def primera_ultima(palabras):
    ordenadas = sorted(palabras, key=str.lower)
    return ordenadas[0], ordenadas[-1]

minimo, maximo = min_max([4, -2, 9, 0])
print(minimo, maximo)
print(primera_ultima(["pera", "Manzana", "uva", "kiwi"]))
```
