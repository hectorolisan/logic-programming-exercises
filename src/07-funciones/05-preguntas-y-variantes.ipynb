## 1. Funciones equivalentes

¿Son equivalentes estas versiones de `mayoria_de_edad`? Sí: las tres devuelven `True` si la edad es al menos 18.

```python
def mayoria_1(edad):
    if edad < 18:
        return False
    else:
        return True

def mayoria_2(edad):
    if edad < 18:
        return False
    return True

def mayoria_3(edad):
    return edad >= 18

print(all(mayoria_1(e) == mayoria_2(e) == mayoria_3(e) for e in range(-5, 100)))
```

## 3. Números perfectos: error y mejora

¿En qué se equivoca este `es_perfecto`? Inicializa `sumatorio` dentro del bucle y devuelve en la primera iteración. Corregirlo y hacerlo más rápido sumando divisores por parejas hasta $\sqrt n$.

```python
from math import isqrt

def es_perfecto_mal(n):
    for i in range(1, n):
        sumatorio = 0
        if n % i == 0:
            sumatorio += i
        return sumatorio == n

def es_perfecto(n):
    if n < 2:
        return False
    total = 1
    for d in range(2, isqrt(n) + 1):
        if n % d == 0:
            total += d + (n // d if d != n // d else 0)
    return total == n

print(es_perfecto_mal(6), [n for n in range(1, 10000) if es_perfecto(n)])
```

## 4. Traza del máximo de una lista

Trazar `maximo([6, 2, 7, 1, 10, 1, 0])` mostrando el candidato en cada paso.

```python
def maximo(lista):
    candidato = lista[0]
    for elemento in lista[1:]:
        if elemento > candidato:
            candidato = elemento
        print(f"  elemento={elemento:>2} candidato={candidato}")
    return candidato

print(maximo([6, 2, 7, 1, 10, 1, 0]))
```

## 5. Funciones sobre listas de números

Media (0 para la lista vacía), productorio, máxima diferencia entre consecutivos (`[1, 10, 2, 6, 2, 0]` → 9) y una versión que devuelve 0 en cuanto encuentra un 0.

```python
from math import prod

def media(lista):
    return sum(lista) / len(lista) if lista else 0

def productorio(lista):
    return prod(lista)

def max_dif_consecutivos(lista):
    return max((abs(a - b) for a, b in zip(lista, lista[1:])), default=0)

def max_dif_o_cero(lista):
    mejor = 0
    for a, b in zip(lista, lista[1:]):
        if a == 0 or b == 0:
            return 0
        mejor = max(mejor, abs(a - b))
    return mejor

print(media([]), media([1, 2, 3]), productorio([1, 2, 3, 4]))
print(max_dif_consecutivos([1, 10, 2, 6, 2, 0]), max_dif_o_cero([1, 10, 2, 6, 2, 0]), max_dif_o_cero([1, 10, 2]))
```

## 6. Pequeñas funciones

Repetir 6 veces una cadena, prefijo común más largo de una lista (`'pol'`), perímetro de un triángulo, logaritmo en base $b$, raíz n-ésima y $\prod_{i=a}^{b} i$ (0 si $a > b$, y 0 sin iterar si el 0 está en el intervalo).

```python
from math import log

def seis_veces(x):
    return x * 6

def prefijo_comun(cadenas):
    prefijo = cadenas[0]
    for c in cadenas[1:]:
        while not c.startswith(prefijo):
            prefijo = prefijo[:-1]
    return prefijo

def perimetro(a, b, c):
    return a + b + c

def log_base(b, x):
    return log(x) / log(b)

def raiz_n_esima(x, n):
    return x ** (1 / n)

def productorio_intervalo(a, b):
    if a > b or a <= 0 <= b:
        return 0
    return prod(range(a, b + 1))

print(seis_veces("ab"), prefijo_comun(["poliedro", "policía", "polífona", "polinizar", "polaridad", "política"]))
print(perimetro(3, 4, 5), log_base(2, 1024), raiz_n_esima(27, 3), productorio_intervalo(3, 6), productorio_intervalo(-2, 4))
```

## 7. Menús con funciones

¿Funciona un `menu` que repite mientras `len(opcion) != 1 or opcion not in 'abc'`? Sí: sin la comprobación de longitud, `'ab'` se aceptaría por ser subcadena. Diseñar `menu_generico(opciones)` y `si_o_no(pregunta)`.

```python
def menu_generico(opciones):
    while True:
        for i, opcion in enumerate(opciones, 1):
            print(f"{i}) {opcion}")
        eleccion = input("Escoja opción: ")
        if eleccion.isdigit() and 1 <= int(eleccion) <= len(opciones):
            return opciones[int(eleccion) - 1]
        print("Opción no válida")

def si_o_no(pregunta):
    while (r := input(pregunta + " ")) not in ("si", "s", "Si", "SI", "no", "n", "No", "NO", "S", "N"):
        print("Responde si, s, Si, SI, no, n, No o NO")
    return r.lower().startswith("s")

print(menu_generico(["Saludar", "Despedirse", "Salir"]))
print(si_o_no("¿Continuar?"))
```

## 10. Alumnos y notas

Con una lista de alumnos y otra de notas: mostrar los aprobados, contar aprobados, mostrar los de nota máxima y los de nota igual o superior a la media, y consultar la nota de un alumno (`None` si no está).

```python
alumnos = ["Ana", "Luis", "Eva", "Pau", "Iker"]
notas = [7.5, 4.0, 9.5, 5.0, 9.5]

def muestra_aprobados(alumnos, notas):
    print([a for a, n in zip(alumnos, notas) if n >= 5])

def num_aprobados(notas):
    return sum(n >= 5 for n in notas)

def muestra_mejores(alumnos, notas):
    print([a for a, n in zip(alumnos, notas) if n == max(notas)])

def muestra_sobre_media(alumnos, notas):
    print([a for a, n in zip(alumnos, notas) if n >= sum(notas) / len(notas)])

def nota_de(alumnos, notas, nombre):
    return notas[alumnos.index(nombre)] if nombre in alumnos else None

muestra_aprobados(alumnos, notas)
print(num_aprobados(notas))
muestra_mejores(alumnos, notas)
muestra_sobre_media(alumnos, notas)
print(nota_de(alumnos, notas, "Eva"), nota_de(alumnos, notas, "Zoe"))
```

## 12. Vuelta ciclista

Con la lista de ciclistas y la matriz de tiempos por etapa: ganador de la vuelta (menor tiempo total), ganador de una etapa y procedimiento que muestre el ganador de cada etapa.

```python
ciclistas = ["Pere Porcar", "Joan Beltran", "Lledó Fabra"]
tiempos = [[10092.0, 12473.1, 13732.3, 10232.1, 10332.3],
           [11726.2, 11161.2, 12272.1, 11292.0, 12534.0],
           [10193.4, 10292.1, 11712.9, 10133.4, 11632.0]]

def ganador_vuelta(ciclistas, tiempos):
    return min(zip(ciclistas, tiempos), key=lambda ct: sum(ct[1]))[0]

def ganador_etapa(ciclistas, tiempos, etapa):
    return min(zip(ciclistas, tiempos), key=lambda ct: ct[1][etapa])[0]

def muestra_ganadores_etapas(ciclistas, tiempos):
    for etapa in range(len(tiempos[0])):
        print(f"Etapa {etapa + 1}: {ganador_etapa(ciclistas, tiempos, etapa)}")

print(ganador_vuelta(ciclistas, tiempos))
muestra_ganadores_etapas(ciclistas, tiempos)
```

## 15. Filtros sobre listas

Funciones que devuelvan los impares de una lista, los nombres que empiezan por una letra, los primos y las parejas de números amigos formadas con un elemento de cada lista.

```python
def suma_divisores(n):
    return sum(d for d in range(1, n // 2 + 1) if n % d == 0)

def impares(lista):
    return [x for x in lista if x % 2]

def empiezan_por(nombres, letra):
    return [n for n in nombres if n.lower().startswith(letra.lower())]

def primos(lista):
    return [x for x in lista if es_primo(x)]

def parejas_amigas(a, b):
    return [(x, y) for x in a for y in b if x != y and suma_divisores(x) == y and suma_divisores(y) == x]

print(impares(range(10)), empiezan_por(["Ana", "Luis", "andrés"], "a"), primos(range(30)))
print(parejas_amigas([220, 1184, 10], [284, 1210, 5]))
```

## 16. Paso de parámetros

¿Qué muestran estos programas? Los enteros se pasan como valores que la función no puede cambiar fuera; las listas se pasan por referencia, salvo que dentro se reasigne el parámetro (`y = y + [4]` crea una lista nueva).

```python
def incrementa(a):
    a = a + 1
    return a

a = 1
b = incrementa(a)
print("a:", a, "b:", b)

def modifica(a, b):
    for elemento in b:
        a.append(elemento)
    b = b + [4]
    a[-1] = 100
    del b[0]
    return b[:]

lista1, lista2 = [1, 2, 3], [1, 2, 3]
lista3 = modifica(lista1, lista2)
print(lista1, lista2, lista3)

def modifica_parametros(x, y):
    x = 1
    y[0] = 1

a, b = 0, [0, 1, 2]
modifica_parametros(a, b)
print(a, b)

def modifica_parametros_2(x, y):
    x = 1
    y.append(3)
    y = y + [4]
    y[0] = 10

a, b = 0, [0, 1, 2]
modifica_parametros_2(a, b)
print(a, b)

def intento_de_intercambio(a, b):
    a, b = b, a

lista1, lista2 = [1, 2], [3, 4]
intento_de_intercambio(lista1, lista2)
print(lista1, lista2)
```

## 17. Invertir una lista en el sitio

¿Invierte bien la lista una función que intercambia `lista[i]` con `lista[-i-1]` para `i` en `range(len(lista) // 2)`? Sí; con longitud impar el elemento central se queda en su sitio. ¿Y `abs_lista`, que modifica la lista recibida?

```python
def invierte(lista):
    for i in range(len(lista) // 2):
        lista[i], lista[-i - 1] = lista[-i - 1], lista[i]

def abs_lista(lista):
    for i in range(len(lista)):
        lista[i] = abs(lista[i])

par, impar = [1, 2, 3, 4], [1, 2, 3, 4, 5]
invierte(par)
invierte(impar)
milista = [1, -1, 2, -3, -2, 0]
abs_lista(milista)
print(par, impar, milista)
```

## 18. Procedimientos frente a funciones

Procedimiento que deje solo los números perfectos; `duplica` (modifica la lista) y `duplica_copia` (no la modifica); y funciones que devuelvan, sin modificar la original, la lista concatenada consigo misma, invertida, concatenada con su inversa y ordenada; y un procedimiento que la ordene en el sitio.

```python
def solo_perfectos(lista):
    lista[:] = [x for x in lista if es_perfecto(x)]

def duplica(lista):
    for i in range(len(lista)):
        lista[i] *= 2

def duplica_copia(lista):
    return [2 * x for x in lista]

def doble(lista):
    return lista + lista

def invertida(lista):
    return lista[::-1]

def con_inversa(lista):
    return lista + lista[::-1]

def ordenada(lista):
    return sorted(lista)

def ordena(lista):
    lista.sort()

l = [6, 3, 28, 10, 496]
solo_perfectos(l)
print(l)
original = [3, 1, 2]
copia = duplica_copia(original)
print(original, copia, doble(original), invertida(original), con_inversa(original), ordenada(original))
duplica(original)
ordena(original)
print(original)
```

## 19. Un único punto de retorno

Una función devolvía la lista `[0, 0, 0]` en dos sitios distintos (por ejemplo, para una lista vacía y para datos no numéricos). Reescribirla para que esa lista solo aparezca una vez.

```python
def resumen(lista):
    """Devuelve [mínimo, máximo, media] o [0, 0, 0] si no se puede calcular."""
    resultado = [0, 0, 0]
    if lista and all(isinstance(x, (int, float)) for x in lista):
        resultado = [min(lista), max(lista), sum(lista) / len(lista)]
    return resultado

print(resumen([3, 1, 2]), resumen([]), resumen([1, "a"]))
```

## 20. Exponencial, seno y coseno con series

Aproximar $e^a = \sum_{k=0}^{n-1} \frac{a^k}{k!}$ también para $a$ negativo ($e^{-a} = 1/e^a$). ¿Es correcta una versión cuyo `factorial` usa `range(2, k)`? No: calcula $(k-1)!$. Aproximar seno y coseno con sus series y hacerlo eficiente calculando cada término a partir del anterior.

```python
from math import cos, exp, sin

def exponencial(a, n=30):
    if a < 0:
        return 1 / exponencial(-a, n)
    termino, suma = 1.0, 0.0
    for k in range(n):
        suma += termino
        termino *= a / (k + 1)
    return suma

def factorial_mal(k):
    p = 1.0
    for i in range(2, k):
        p *= i
    return p

def seno(x, n=20):
    termino, suma = x, 0.0
    for k in range(n):
        suma += termino
        termino *= -x * x / ((2 * k + 2) * (2 * k + 3))
    return suma

def coseno(x, n=20):
    termino, suma = 1.0, 0.0
    for k in range(n):
        suma += termino
        termino *= -x * x / ((2 * k + 1) * (2 * k + 2))
    return suma

print(exponencial(2), exp(2), exponencial(-3), exp(-3))
print(factorial_mal(5), "≠ 120")
print(seno(1), sin(1), coseno(1), cos(1))
```

## 21. Bisección

Encontrar un cero de $f$ en $[a, b]$ por bisección. Si $f(a)$ y $f(b)$ tienen el mismo signo, devolver `None`; si alguno vale 0, devolver ese extremo. Pedir $a < b$ y la precisión $\varepsilon$ al usuario.

```python
def biseccion(f, a, b, epsilon):
    if f(a) == 0:
        return a
    if f(b) == 0:
        return b
    if f(a) * f(b) > 0:
        return None
    while b - a > epsilon:
        c = (a + b) / 2
        if f(a) * f(c) <= 0:
            b = c
        else:
            a = c
    return (a + b) / 2

f = lambda x: x**3 - 2 * x - 5
a, b = float(input("a: ")), float(input("b: "))
while a >= b:
    print("a debe ser menor que b")
    a, b = float(input("a: ")), float(input("b: "))
epsilon = float(input("epsilon: "))
print(biseccion(f, a, b, epsilon), biseccion(f, 0, 1, epsilon))
```
