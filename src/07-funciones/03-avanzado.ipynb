## 1. Número de series

Una *serie* es un tramo de valores idénticos consecutivos. Escribir una función que cuente las series de una lista (`[1, 1, 8, 8, 8, 8, 0, 0, 0, 2, 10, 10]` tiene 5).

```python
def num_series(lista):
    return sum(1 for i, v in enumerate(lista) if i == 0 or v != lista[i - 1])

print(num_series([1, 1, 8, 8, 8, 8, 0, 0, 0, 2, 10, 10]))
```

## 2. Serie más larga

Escribir una función que devuelva la posición donde empieza la serie más larga (la primera si hay empate). En el ejemplo anterior es 2; para `[8, 2, 2, 9, 9]` es 1.

```python
def inicio_serie_mas_larga(lista):
    mejor_inicio, mejor_largo, inicio = 0, 0, 0
    for i in range(1, len(lista) + 1):
        # cierra serie al cambiar valor o al final
        if i == len(lista) or lista[i] != lista[inicio]:
            if i - inicio > mejor_largo:
                mejor_inicio, mejor_largo = inicio, i - inicio
            inicio = i
    return mejor_inicio

print(inicio_serie_mas_larga([1, 1, 8, 8, 8, 8, 0, 0, 0, 2, 10, 10]), inicio_serie_mas_larga([8, 2, 2, 9, 9]))
```

## 3. Máxima diferencia

Escribir una función que devuelva el valor absoluto de la máxima diferencia entre cualquier par de elementos de una lista (`[1, 10, 2, 6, 8, 2, 0]` → 9).

```python
def maxima_diferencia(lista):
    return max(lista) - min(lista)

print(maxima_diferencia([1, 10, 2, 6, 8, 2, 0]))
```

## 4. Números amigos

Dos números son amigos si la suma de los divisores propios de cada uno es igual al otro. Escribir una función que diga si dos números son amigos y un procedimiento que muestre las parejas de amigos menores que $n$, sin repetir parejas.

```python
def suma_divisores(n):
    total = 1
    for d in range(2, int(n**0.5) + 1):
        if n % d == 0:
            total += d + (n // d if d != n // d else 0)
    return total

def son_amigos(a, b):
    return a != b and suma_divisores(a) == b and suma_divisores(b) == a

def muestra_amigos(n):
    for a in range(2, n):
        b = suma_divisores(a)
        # a < b evita parejas repetidas
        if a < b < n and son_amigos(a, b):
            print(f"{a} y {b} son amigos")

print(son_amigos(220, 284))
muestra_amigos(10000)
```

## 5. Cifras en orden inverso

Escribir un procedimiento que muestre las cifras de un entero en orden inverso, cada una en una línea, usando operaciones aritméticas.

```python
def cifras_inversas(n):
    n = abs(n)
    while True:
        n, cifra = divmod(n, 10)
        print(cifra)
        if n == 0:
            break

cifras_inversas(324)
```

## 6. Operaciones de conjuntos con listas

Escribir funciones que reciban dos listas y devuelvan, sin repetidos, su intersección, su unión y su diferencia (sin usar `set`).

```python
def sin_repetidos(lista):
    resultado = []
    for x in lista:
        if x not in resultado:
            resultado.append(x)
    return resultado

def interseccion(a, b):
    return sin_repetidos([x for x in a if x in b])

def union(a, b):
    return sin_repetidos(a + b)

def diferencia(a, b):
    return sin_repetidos([x for x in a if x not in b])

a, b = [1, 2, 1], [2, 3, 2, 4]
print(interseccion(a, b), union(a, b), diferencia(a, b))
```

## 7. Todos los pares

Escribir una función que, dadas dos listas, devuelva todos los pares formados por un elemento de la primera y otro de la segunda (`[1, 3, 5]` y `[2, 5]` → `[[1, 2], [1, 5], [3, 2], ...]`).

```python
def pares(a, b):
    return [[x, y] for x in a for y in b]

print(pares([1, 3, 5], [2, 5]))
```

## 8. Lista y su inversa

Escribir una función que devuelva una copia de la lista concatenada con su inversa.

```python
def con_inversa(lista):
    return lista + lista[::-1]

print(con_inversa([1, 2, 3]))
```

## 9. Traza de una matriz

Escribir una función que, si la matriz es cuadrada, devuelva la suma de su diagonal principal y, si no, `None`.

```python
def traza(M):
    if any(len(fila) != len(M) for fila in M):
        return None
    return sum(M[i][i] for i in range(len(M)))

print(traza([[1, 2], [3, 4]]), traza([[1, 2, 3]]))
```

## 10. Ecuación de segundo grado

Escribir una función que reciba $a$, $b$ y $c$ de $ax^2 + bx + c = 0$ y devuelva una lista con sus dos soluciones reales; si solo hay una, dos copias de ella; si no hay o son infinitas, `[None, None]`.

```python
from math import sqrt

def segundo_grado(a, b, c):
    if a == 0:
        return [-c / b] * 2 if b != 0 else [None, None]
    d = b**2 - 4 * a * c
    if d < 0:
        return [None, None]
    return [(-b + sqrt(d)) / (2 * a), (-b - sqrt(d)) / (2 * a)]

print(segundo_grado(1, -3, 2), segundo_grado(1, 2, 1), segundo_grado(1, 0, 1), segundo_grado(0, 0, 0))
```
