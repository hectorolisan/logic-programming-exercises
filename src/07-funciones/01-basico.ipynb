## 1. Saludo

Escribir una función que muestre `¡Hola amiga!` cada vez que se invoque.

```python
def saluda():
    print("¡Hola amiga!")

saluda()
saluda()
```

## 2. Saludo con nombre

Escribir una función que reciba un nombre y muestre `¡hola <nombre>!`.

```python
def saluda_a(nombre):
    print(f"¡hola {nombre}!")

saluda_a("Carmen")
```

## 3. Factorial

Escribir una función que reciba un entero positivo y devuelva su factorial.

```python
def factorial(n):
    resultado = 1
    for i in range(2, n + 1):
        resultado *= i
    return resultado

print(factorial(5), factorial(0))
```

## 4. Factura con IVA

Escribir una función que reciba el importe sin IVA y el porcentaje de IVA (21 % por defecto) y devuelva el total.

```python
def total_factura(importe, iva=21):
    return round(importe * (1 + iva / 100), 2)

print(total_factura(100), total_factura(100, 10))
```

## 5. Círculo y cilindro

Escribir una función que calcule el área de un círculo y otra que calcule el volumen de un cilindro usando la primera.

```python
from math import pi

def area_circulo(radio):
    return pi * radio**2

def volumen_cilindro(radio, altura):
    return area_circulo(radio) * altura

print(area_circulo(1), volumen_cilindro(1, 10))
```

## 6. Media

Escribir una función que reciba una lista de números y devuelva su media.

```python
def media(muestra):
    return sum(muestra) / len(muestra)

print(media([1, 2, 3, 4]))
```

## 7. Cuadrados

Escribir una función que reciba una lista de números y devuelva otra con sus cuadrados.

```python
def cuadrados(muestra):
    return [x**2 for x in muestra]

print(cuadrados([1, 2, 3, 4]))
```

## 8. Conversiones

Definir funciones para: raíz cúbica, área de un círculo a partir del radio, Fahrenheit → Celsius, Celsius → Fahrenheit, radianes → grados y grados → radianes.

```python
from math import pi

def raiz_cubica(x):
    # 1 / 3 es división real, no entera
    return x ** (1 / 3)

def fahrenheit_a_celsius(f):
    return (f - 32) * 5 / 9

def celsius_a_fahrenheit(c):
    return c * 9 / 5 + 32

def radianes_a_grados(r):
    return r * 180 / pi

def grados_a_radianes(g):
    return g * pi / 180

print(raiz_cubica(27), fahrenheit_a_celsius(212), celsius_a_fahrenheit(37))
print(radianes_a_grados(pi), grados_a_radianes(90))
```

## 9. Suma de argumentos arbitrarios

Escribir `add_all_nums`, que reciba cualquier número de argumentos y los sume, avisando si alguno no es numérico.

```python
def add_all_nums(*nums):
    no_numericos = [n for n in nums if not isinstance(n, (int, float)) or isinstance(n, bool)]
    if no_numericos:
        return f"No son números: {no_numericos}"
    return sum(nums)

print(add_all_nums(1, 2, 3.5), add_all_nums(1, "2", None))
```

## 10. Funciones sobre listas

Escribir `print_list` (muestra cada elemento), `reverse_list` (invierte con un bucle), `capitalize_list_items`, `add_item` (añade al final) y `remove_item` (elimina un elemento).

```python
def print_list(lista):
    for x in lista:
        print(x)

def reverse_list(lista):
    invertida = []
    for x in lista:
        invertida = [x] + invertida
    return invertida

def capitalize_list_items(lista):
    return [x.capitalize() for x in lista]

def add_item(lista, item):
    return lista + [item]

def remove_item(lista, item):
    return [x for x in lista if x != item]

print_list(["a", "b"])
print(reverse_list([1, 2, 3, 4, 5]), capitalize_list_items(["potato", "milk"]))
print(add_item(["Potato", "Tomato", "Mango", "Milk"], "Meat"), remove_item([2, 3, 7, 9], 3))
```

## 11. Sumas hasta n

Escribir `sum_of_numbers(n)`, `sum_of_odds(n)` y `sum_of_even(n)`, que sumen todos los números, los impares y los pares de 0 a $n$.

```python
def sum_of_numbers(n):
    return sum(range(n + 1))

def sum_of_odds(n):
    return sum(range(1, n + 1, 2))

def sum_of_even(n):
    return sum(range(0, n + 1, 2))

print(sum_of_numbers(100), sum_of_odds(100), sum_of_even(100))
```
