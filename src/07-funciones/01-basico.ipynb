## 1. Saludo

Escribir una función que muestre `¡Hola amiga!` cada vez que se invoque.

```python
def saluda():
    print("¡Hola amiga!")

saluda()
saluda()
```

## 2. Saludo con nombre

Escribir una función que reciba un nombre y muestre `¡hola <nombre>!`.

```python
def saluda_a(nombre):
    print(f"¡hola {nombre}!")

saluda_a("Carmen")
```

## 3. Factorial

Escribir una función que reciba un entero positivo y devuelva su factorial.

```python
def factorial(n):
    resultado = 1
    for i in range(2, n + 1):
        resultado *= i
    return resultado

print(factorial(5), factorial(0))
```

## 4. Factura con IVA

Escribir una función que reciba el importe sin IVA y el porcentaje de IVA (21 % por defecto) y devuelva el total.

```python
def total_factura(importe, iva=21):
    return round(importe * (1 + iva / 100), 2)

print(total_factura(100), total_factura(100, 10))
```

## 5. Círculo y cilindro

Escribir una función que calcule el área de un círculo y otra que calcule el volumen de un cilindro usando la primera.

```python
from math import pi

def area_circulo(radio):
    return pi * radio**2

def volumen_cilindro(radio, altura):
    return area_circulo(radio) * altura

print(area_circulo(1), volumen_cilindro(1, 10))
```

## 6. Media

Escribir una función que reciba una lista de números y devuelva su media.

```python
def media(muestra):
    return sum(muestra) / len(muestra)

print(media([1, 2, 3, 4]))
```

## 7. Cuadrados

Escribir una función que reciba una lista de números y devuelva otra con sus cuadrados.

```python
def cuadrados(muestra):
    return [x**2 for x in muestra]

print(cuadrados([1, 2, 3, 4]))
```

## 8. Conversiones

Definir funciones para: raíz cúbica, área de un círculo a partir del radio, Fahrenheit → Celsius, Celsius → Fahrenheit, radianes → grados y grados → radianes.

```python
from math import pi

def raiz_cubica(x):
    # 1 / 3 es división real, no entera
    return x ** (1 / 3)

def fahrenheit_a_celsius(f):
    return (f - 32) * 5 / 9

def celsius_a_fahrenheit(c):
    return c * 9 / 5 + 32

def radianes_a_grados(r):
    return r * 180 / pi

def grados_a_radianes(g):
    return g * pi / 180

print(raiz_cubica(27), fahrenheit_a_celsius(212), celsius_a_fahrenheit(37))
print(radianes_a_grados(pi), grados_a_radianes(90))
```
